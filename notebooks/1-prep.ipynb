## Anaconda installieren

Im Grunde bräuchten wir nur eine Python-Installation und einen Texteditor zum Programmieren. Wir gönnen uns aber etwas mehr Komfort und installieren "Anaconda", eine umfassende Python-Distribution, die viele nützliche Bibliotheken und Werkzeuge integriert. Mit Anaconda können wir Python und wichtige Pakete problemlos installieren, verwalten und aktualisieren. Die integrierte Benutzeroberfläche, der Anaconda Navigator, ermöglicht den einfachen Zugriff auf beliebte Entwicklungsumgebungen wie Jupyter Notebooks oder Visual Studio Code (siehe unten). 

Installationsanweisungen für die Plattformen Win / macOS / Linux findet man [hier](https://www.anaconda.com/docs/getting-started/anaconda/install).

Mit der Anaconda-Installation wird auch der Anaconda Navigator installiert, der eine grafische Benutzeroberfläche zur Verfügung stellt, mit der man Conda-Umgebungen und Bibliotheken verwalten sowie Programme (z.B. IDEs) starten kann. Eine Dokumentation des Navigators findet man [hier](https://www.anaconda.com/docs/tools/anaconda-navigator/main), die Bedienung ist aber intuitiv, probieren Sie es aus! 

::: {.callout-important}
### Ist Anaconda kostenlos? {.unnumbered}
Für Studierende und Lehrende an Hochschulen: ja. Anaconda erlaubt die kostenlose Nutzung für nicht-kommerzielle Zwecke in Lehre, Studium und Forschung an akkreditierten Hochschulen (sog. *Academic Policy*). In Unternehmen mit 200 oder mehr Beschäftigten ist für die Anaconda-Paketquellen dagegen eine kostenpflichtige Lizenz nötig. Eine für jeden Zweck kostenlose Alternative ist **Miniforge**: Es enthält Python und den Paketmanager `conda`, bezieht die Pakete aber aus der freien Paketquelle *conda-forge*. Eine grafische Oberfläche wie den Navigator bringt Miniforge allerdings nicht mit.
:::

::: {.callout-note}
### Was ist eine Umgebung in Anaconda? {.unnumbered}
Umgebungen (engl. *environments*) sind isolierte Arbeitsbereiche, in denen man verschiedene Versionen von Python und Paketen installieren und verwalten kann, ohne dass diese sich gegenseitig beeinflussen. Dies ist besonders nützlich, wenn man an mehreren Projekten arbeitet, die unterschiedliche Abhängigkeiten und Versionen von Bibliotheken erfordern. Die Standardumgebung heißt "base" und wir benötigen in diesem Kurs keine weiteren Umgebungen (für eigene größere Projekte empfiehlt es sich später, je Projekt eine eigene Umgebung anzulegen). Zum Experimentieren können Sie sich aber eine weitere Umgebung anlegen und dort Pakete installieren und testen, ohne Sorge zu haben, in Ihrer Standardumgebung etwas "kaputt zu machen".
:::

## Die IDE - unsere Werkbank fürs Programmieren!

Beim Programmieren ist eine integrierte Entwicklungsumgebung (IDE, engl.: integrated development environment) das wichtigste Werkzeug – vergleichbar mit der Werkbank eines Handwerkers. Eine IDE kombiniert verschiedene Funktionen, die das Coden effizienter und angenehmer machen. Über Anaconda können wir verschiedene IDEs einfach über den Navigator installieren. Aber auch IDEs, die außerhalb des Navigators installiert werden, können in Anaconda integriert und später über den Navigator gestartet werden. 

### Was ist eine IDE? {.unnumbered}
Eine IDE ist eine Software, die mehrere Programmier-Tools in einer einzigen Oberfläche bündelt. Typische Bestandteile sind:

* Code-Editor mit Syntax-Highlighting und Autovervollständigung. Das ist das Herzstück: Ein Texteditor, mit dem wir unseren Quellcode schreiben.
* Debugger, um Fehler im Code zu finden. Beim "Debuggen" sucht man systematisch nach einem Fehler im Programm und nutzt hierzu verschiedene technische Hilfsmittel (z.B. "Breakpoints", die das Programm an bestimmten Stellen anhalten, um Variablen gezielt auszulesen und das gewünschte Verhalten zu überprüfen).
* Anbindung an einen Interpreter oder Compiler, um den Code auszuführen (die IDE nutzt dafür meist eine separat installierte Python-Version, z.B. aus Anaconda).
* Versionskontrolle, um Änderungen am Code zu verwalten.

### Vorteile einer IDE {.unnumbered}
Warum sollte man eine IDE nutzen? Die Alternative zur IDE wäre ein einfacher Texteditor ohne spezielle Coding-Funktionalitäten. Hier sind einige Gründe, die stark für die IDE sprechen:

✅ Produktivität: Schnelleres Schreiben, Testen und Debuggen von Code.

✅ Fehlerminimierung: Automatische Code-Analyse und Debugging-Werkzeuge.

✅ Komfort: Bessere Übersicht und Organisation durch Projektverwaltung.

### IDEs für die Python-Entwicklung {.unnumbered}

1. **PyCharm**

    PyCharm ist eine der bekanntesten IDEs für Python und wird von JetBrains entwickelt. Seit 2025 gibt es nur noch *ein* PyCharm: Die Grundfunktionen sind kostenlos – einschließlich der Unterstützung für Jupyter Notebooks –, zusätzliche Profi-Funktionen erfordern ein kostenpflichtiges Pro-Abonnement. Die frühere kostenlose Community Edition wurde damit eingestellt. Studierende erhalten die Pro-Funktionen kostenlos über das *JetBrains Student Pack* (jeweils für ein Jahr, verlängerbar; siehe [PyCharm für Studierende](https://lp.jetbrains.com/pycharm-for-students/)).

   Getting Started with PyCharm: [Guide](https://www.jetbrains.com/help/pycharm/getting-started.html). 

2. **Visual Studio Code**

    Visual Studio Code (VS Code) von Microsoft ist ein beliebter, kostenloser Code-Editor, der durch Erweiterungen (Extensions) zu einer vollwertigen IDE für viele Programmiersprachen wird. Für Python installiert man die Erweiterung *Python* von Microsoft, für Notebooks zusätzlich die Erweiterung *Jupyter* (siehe unten). Hier geht es zur [Dokumentation](https://code.visualstudio.com/docs).


3. **Jupyter Notebooks**


   Jupyter Notebook ist eine webbasierte Entwicklungsumgebung, die besonders beliebt für interaktive Datenanalyse oder maschinelles Lernen ist. Das Besondere ist die Kombination aus Markdown- und Code-Zellen. Während in Code-Zellen programmiert wird, können in Markdown-Zellen Texte formatiert werden, aber auch Bilder oder Animationen eingebunden werden (Markdown-Zellen unterstützen auch LaTeX-Befehle zur Darstellung mathematischer Formeln). So entsteht ein interaktives Skript, welches mehr als reiner Quellcode ist. Wir werden in der Vorlesung oft Jupyter Notebooks verwenden, da die Kombination aus Text, Bildern und Code für die Lehre sehr vorteilhaft ist. Auch dieses Skript ist in Jupyter Notebooks entstanden. Mehr darüber gibt es [hier](https://docs.jupyter.org/en/latest/). 

    Zur Begriffsklärung: *Notebook* bezeichnet einerseits das Dateiformat (`.ipynb`), andererseits das Programm *Jupyter Notebook*, mit dem man solche Dateien im Browser bearbeitet. Daneben gibt es **JupyterLab**, eine umfangreichere Oberfläche mit Dateibrowser und mehreren Tabs. Seit Version 7 basiert auch Jupyter Notebook technisch auf JupyterLab. Beide öffnen dieselben Dateien, und auch VS Code und PyCharm können `.ipynb`-Dateien direkt bearbeiten.


4. **Spyder**

    Spyder ist eine IDE speziell für wissenschaftliches Rechnen (aktuell Version 6) und kann direkt über den Anaconda Navigator gestartet bzw. installiert werden; es gibt sie auch als eigenständiges Installationsprogramm. Ihr *Variable Explorer* zeigt alle Variablen und Arrays übersichtlich als Tabelle an. Wer MATLAB kennt, fühlt sich hier schnell zu Hause. Mehr dazu [hier](https://www.spyder-ide.org/).


5. **Cloud-Alternative: Google Colab**


   Google Colab (Colaboratory) ist eine cloudbasierte Entwicklungsumgebung. Mit Google Colab kann Python-Code direkt im Browser geschrieben und ausgeführt werden, ohne dass eine lokale Installation erforderlich ist. Im Prinzip können wir somit auch von einem Tablet oder Smartphone aus programmieren (ein kleiner Bildschirm und eine fehlende Tastatur können dennoch sehr herausfordernd sein beim Coden). Außerdem kann man GPUs und TPUs nutzen, ohne sich selbst Hardware anschaffen zu müssen. In der kostenlosen Version ist der Zugriff darauf allerdings stark eingeschränkt und nicht garantiert, und eine Sitzung läuft höchstens 12 Stunden. Für diesen Kurs brauchen wir keine GPU. In Google Colab arbeitet man ebenfalls mit der beliebten Notebook-Struktur aus Markdown- und Code-Zellen. Wenn man diese nicht braucht, kann man den gesamten Code in eine einzelne Zelle schreiben. Google Colab ermöglicht auch das gemeinsame Arbeiten an einem Projekt oder die Freigabe von Notebooks für andere. Zur Nutzung der Online-IDE ist ein (kostenfreies) Google-Konto erforderlich; beachten Sie, dass Ihre Notebooks dabei in Ihrem Google Drive gespeichert werden. Zu Google Colab geht es [hier](https://colab.research.google.com/).

    Eine Alternative wäre **PythonAnywhere** (siehe [Website](https://www.pythonanywhere.com/)). Dort lassen sich Python-Skripte und kleine Web-Anwendungen im Browser ausführen; GPUs gibt es dort allerdings nicht, auch nicht in den kostenpflichtigen Tarifen.

### KI-Assistenten in der IDE {.unnumbered}
Moderne IDEs lassen sich um KI-Assistenten erweitern, die Code vorschlagen oder erklären. GitHub Copilot ist für verifizierte Studierende über GitHub Education kostenlos. Für die ersten Schritte empfehlen wir allerdings ausdrücklich, auf KI-Unterstützung zu verzichten: Die Grundlagen müssen Sie zunächst selbst beherrschen. Erst dann können Sie KI-Vorschläge sinnvoll bewerten, einordnen und in Ihren eigenen Code einbinden – und genau dafür sind KI-Assistenten später in der Praxis eine große Hilfe.
:::
