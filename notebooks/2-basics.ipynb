## Das erste Programm: Hello, World

Wann immer man eine neue Programmiersprache lernt oder eine Programmierumgebung neu einrichtet (z.B. eine neue IDE installiert und ausprobiert), fängt man mit dem berühmten "Hello, World"-Programm an. Aufgabe des Programms ist die Ausgabe des Textes "Hello, World" auf dem Bildschirm. Ist dies gelungen, weiß man, dass man eine arbeitsfähige Umgebung hat, und kann sich komplizierteren Aufgaben zuwenden. Let's do it:  

````python
print('Hello, World')
````

In Python benutzen wir den Befehl `print` für eine Bildschirmausgabe. Er ist gleichzeitig unser erstes Beispiel für eine Funktion, das erkennen wir an den runden Klammern nach dem Befehl: Hier erwartet die Funktion ein Argument oder mehrere Argumente. Das kennen wir aus der Mathematik: Die Funktion $f(x)$ erwartet als Argument einen Wert $x$. Funktionen in der Programmierung unterscheiden sich allerdings in einigen Punkten von Funktionen in der Mathematik, wir werden z.B. gleich sehen, dass `print` mit beliebig vielen Argumenten aufgerufen werden kann, während $f(x)$ immer genau ein Argument erwartet.

Die Funktion `print(...)` wird mit dem auszugebenden Text aufgerufen. Texte werden in Python in einfache (`'`) oder doppelte (`"`) Anführungszeichen gesetzt. Wir hätten also auch diesen Befehl verwenden können:  

````python
print("Hello, World")
````

Der `print`-Befehl in Python ist ein grundlegender und oft verwendeter Befehl, der dazu dient, Texte oder andere Daten auf dem Bildschirm anzuzeigen. Das ist sehr nützlich, um Informationen während der Ausführung eines Programms zu überprüfen oder Ergebnisse von Berechnungen anzuzeigen.

Wie bereits erwähnt, kann man auch mehr als ein Argument übergeben (getrennt durch ein Komma):

````python
print("Hello,", "World")
````

Das Ergebnis ist dasselbe, da die einzelnen Argumente standardmäßig durch ein Leerzeichen getrennt werden. Dieses Verhalten kann durch weitere Argumente geändert werden. Probieren Sie das Folgende einmal aus:

````python
#| output: false
print("It is", "a beautiful", "day.", sep = "-")
````

Nun haben Sie schon ein sog. „Keyword-Argument“ kennengelernt: `sep`.
Ein Keyword-Argument erkennen Sie daran, dass es explizit einem Namen zugewiesen wird, in unserem Beispiel durch `sep=...`. (Das Gegenstück dazu sind die sog. „Positionsargumente“: alle Argumente, die ohne Zuweisung übergeben werden. Ohne das Keyword `sep` könnte die `print`-Funktion das Argument nicht von den vorherigen unterscheiden. Auf den Unterschied zwischen Positions- und Keyword-Argumenten werden wir natürlich später noch ausführlicher eingehen. Wir sind hier noch bei unseren ersten Gehversuchen in der Welt der Programmierung.)

## Zuweisungen

In der Programmierung sind Zuweisungen ein grundlegendes Konzept, um Werte in Variablen zu speichern. Eine Zuweisung erfolgt in Python mit dem Gleichheitszeichen (=). Schauen wir uns Beispiele an:

````python
# Zuweisung:
# Die Variable x erhält den Wert 5
x = 5 
````

````python
print(x)  # Überzeugen wir uns davon!
````

````python
# Die Variable name erhält
# den Wert "Tina"
name = "Tina" 
````

````python
# Die Variable y erhält 
# den Wert 1.3
y = 1.3 
````

Mit Variablen können wir rechnen:

````python
# Die Variable z erhält 
# den Wert der Addition x + y
z = x + y 
````

````python
print(z)  # Überzeugen wir uns davon!
````

::: {.callout-caution}
Das Gleichheitszeichen ist nicht mit dem mathematischen Gleichheitszeichen zu verwechseln: In der Mathematik sind beide Seiten einer Gleichung gleichberechtigt und die Gleichung x = 5 ist entweder wahr oder falsch. In der Programmierung führen wir aber eine Operation durch, nämlich eine Zuweisung: Der Wert auf der rechten Seite wird der Variable auf der linken Seite zugewiesen! 
:::

Machen Sie sich an dem folgenden Beispiel klar, dass `x = y` und `y = x` nicht dasselbe sind in der Programmierung!

````python
# Beispiel:
x = 5
y = 7
x = y # was würde y = x am Ergebnis ändern?
print(x, y) 
````

Die bisherigen Beispiele waren sog. Einfach-Zuweisungen: Einer Variable wurde ein Wert zugewiesen. In Python kann man aber auch mehreren Variablen gleichzeitig verschiedene Werte zuweisen, indem man links mehrere Namen und rechts mehrere Werte angibt, jeweils durch Kommas getrennt (diese Schreibweise gibt es nicht in allen Programmiersprachen, in C zum Beispiel nicht):

````python
# Zwei Variablen 
# gleichzeitig zuweisen:
name, age = "Tina", 25 
````

````python
print("Ich heiße", name, "und bin", age, "Jahre alt.")
````

**Einschub:** Die letzte Ausgabe geht deutlich leichter, wie das folgende Beispiel zeigt.

````python
# Achten Sie auf das kleine f und die geschweiften
# Klammern:
print(f"Ich heiße {name} und bin {age} Jahre alt.")
# Hier verwenden wir einen sog. f-String. 
# Das f steht für formatiert bzw. formatted.
# Es sagt Python, dass die Variablen in den geschweiften 
# Klammern durch ihre Werte ersetzt werden sollen.
````

Bei Zahlen lässt sich in f-Strings zusätzlich die **Darstellung** festlegen. Dazu schreibt man hinter den Variablennamen einen Doppelpunkt und eine Formatangabe, z.B. `.2f` für zwei Nachkommastellen oder `.1e` für die Exponentialschreibweise mit einer Nachkommastelle. Das ist besonders praktisch für Messwerte:

````python
voltage = 3.14159  # Spannung in V
current = 0.0123   # Strom in A

print(f"Spannung: {voltage:.2f} V")         # zwei Nachkommastellen
print(f"Strom: {current:.1e} A")            # Exponentialschreibweise
print(f"Strom: {current * 1000:.1f} mA")    # auch Rechnungen sind möglich
````

Beachten Sie: Die Formatangabe ändert nur die *Darstellung* in der Ausgabe – der Wert der Variable bleibt unverändert.

Selbstverständlich kann sich der Wert einer Variable nach der ersten Zuweisung noch verändern, indem ein neuer Wert zugewiesen wird. Das demonstrieren wir anhand verschiedener Beispiele:

* **Beispiel mit einer einzelnen Variable.**

````python
x = 20  # x hat den Wert 20
x = 21  # x hat den Wert 21
x = 22  # x hat den Wert 22
print(x)  # Überzeugen wir uns!
````

* **Beispiel mit zwei Variablen.**

````python
x = 1.2  # x hat den Wert 1.2
y = 42  # y hat den Wert 42
x = y  # x bekommt den Wert von y, also 42
y = 0  # y bekommt den Wert 0, x bleibt unverändert
print(x, y)
````

* **Beispiel, welches den Unterschied zur mathematischen Gleichung noch einmal deutlich hervorhebt.**

````python
x = 2
x = x + 2 # Das Ergebnis von x+2 wird der Variable x zugewiesen.
print(x)
````

::: {.callout-note}
### Erweiterte Zuweisungen {.unnumbered}
Mit `+=`, `-=`, `*=`, `/=` können Zuweisungen wie im letzten Beispiel verkürzt werden. Dabei steht `x += 5` für `x = x + 5` (die anderen analog). Es wird also eine mathematische Operation mit der Variable durchgeführt (z.B. "addiere 5") und das Ergebnis derselben Variablen als neuer Wert zugewiesen. Es handelt sich lediglich um eine kürzere Schreibweise. Für Zahlen und Strings (Zeichenketten) ist das genau gleichwertig. Bei Listen gibt es einen kleinen, aber wichtigen Unterschied, den wir im Kapitel *Datenstrukturen* kennenlernen. 
:::

* **Beispiel mit erweiterten Zuweisungen.**

````python
x, y, z = 1, 2, 3
x += 10
y -= 10
z *= 10
print(x, y, z)
````

**Wichtig:** Python-Code wird vom Interpreter der Reihe nach abgearbeitet. Wenn Sie eine Variable z.B. erst nach einem Aufruf von `print` verändern, so bekommt der `print`-Befehl das nicht mit. Beispiel:

````python
x = 5
print(x)
x += 10  # für das print darüber unerheblich!
print(x) # jetzt sieht man die Veränderung!
````

### Gute Variablennamen {.unnumbered}
Obwohl Python im Vergleich zu anderen Programmiersprachen mehr Freiheiten bei der Benennung von Variablen zulässt (z.B. die Verwendung von Umlauten und anderen speziellen Buchstaben), sollte man sich auf das lateinische Alphabet beschränken. Der offizielle Python-Style-Guide ([PEP 8](https://peps.python.org/pep-0008/#naming-conventions)) empfiehlt mit einem kleinen Buchstaben anzufangen und mehrere Wörter mit einem Unterstrich (`_`) zu trennen (das nennt man *snake_case*). Es sind auch Ziffern erlaubt (allerdings nicht am Anfang). Wichtig ist eine gute Verständlichkeit, damit auch andere Menschen Ihren Code nachvollziehen können. Daher empfehlen sich kurze, aber aussagekräftige Namen. Es folgen ein paar Beispiele:

`customer_name = "Markus"`\
`user_age = 32`\
`total_price = 3.4e3`\
`coordinate_1, coordinate_2, coordinate_3 = 3.1, -2.0, 0.0`\
`x_coordinate, y_coordinate, z_coordinate = -1.0, -1.0, 1.0`

Im Style-Guide wird außerdem davor gewarnt, dass die Buchstaben 'l' (kleines L), 'O' (großes o) oder 'I' (großes i) leicht mit den Zahlen '1' bzw. '0' verwechselt werden können (in manchen Schriftarten werden sie identisch dargestellt!). Immer darauf achten, ob es aus dem Kontext eindeutig zu erkennen ist.    

Viele Organisationen oder Teams definieren übrigens eigene Style-Guides, die weitere Details festlegen. In internationalen Projekten legt man z.B. häufig fest, dass die englische Sprache im Code zu verwenden ist. Andere Unternehmen verwenden deutsche Wörter. Wenn Sie Teil eines Teams werden, fragen Sie nach den internen Konventionen (explizit niedergeschrieben oder implizit von allen befolgt). 

Abschließend sei an dieser Stelle darauf hingewiesen, dass Sie keine Schlüsselwörter von Python für Ihre Variablennamen verwenden dürfen, z.B. `if`, `else`, `class`, `return` usw. (Python meldet dann einen `SyntaxError`). Namen eingebauter Funktionen wie `print` oder `sum` sind dagegen erlaubt – aber gefährlich (siehe folgenden Kasten).

### Fallstrick: Namen eingebauter Funktionen als Variablennamen {.unnumbered}
Python bringt viele eingebaute Funktionen mit, z.B. `print()` oder `sum()` (addiert alle Zahlen einer Folge, mehr dazu im Kapitel *Funktionen*). Python verbietet nicht, diese Namen für eigene Variablen zu verwenden – mit unangenehmen Folgen:

```python
sum = 3 + 4             # die Variable heißt wie die eingebaute Funktion sum()
print(sum([1, 2, 3]))   # TypeError: 'int' object is not callable
```

Ab der Zuweisung zeigt `sum` nicht mehr auf die Funktion, sondern auf die Zahl 7 – und eine Zahl kann man nicht aufrufen. Der Fehler tritt oft erst viel später im Code auf und ist dann schwer zu finden. Wählen Sie deshalb einen anderen Namen, z.B. `total`. Ist es schon passiert, stellt `del sum` in Jupyter die eingebaute Funktion wieder her (oder Sie starten den Kernel – das Python im Hintergrund des Notebooks – neu).

## Vergleiche

In Programmen wird viel gerechnet, wofür die arithmetischen Operationen, die wir bisher betrachtet haben, die Grundlage bilden. Ein weiteres grundlegendes Element von Programmen ist die **Logik**: Dabei geht es um die Entscheidung, ob etwas wahr oder falsch ist. Je nach Ergebnis kann ein Programm unterschiedliche Wege einschlagen: Ist eine Bedingung wahr, wird eine bestimmte Aktion ausgeführt. Ist sie falsch, passiert etwas anderes.

Schauen wir uns nun die grundlegenden logischen Ausdrücke an, mit denen wir solche Entscheidungen treffen können. Vergleichen wir zunächst zwei Zahlen:

````python
1 < 42  # Ist 1 kleiner als 42?
````

````python
1 > 42  # Ist 1 größer als 42?
````

````python
2 <= 3  # Ist 2 kleiner oder gleich 3?
````

````python
2 >= 3  # Ist 2 größer oder gleich 3?
````

````python
2 == 3  # Ist 2 gleich 3?
````

````python
2 != 3  # Ist 2 ungleich 3?
````

::: {.callout-caution}
Für den logischen Vergleich verwendet man das doppelte Gleichheitszeichen `==`, welches nicht mit dem Zuweisungsoperator `=` verwechselt werden darf! 
:::

Wie Sie festgestellt haben, liefern logische Vergleiche als Ergebnis einen Wahrheitswert: `True` für wahr oder `False` für falsch. Wahrheitswerte können genauso wie andere Werte einer Variable zugewiesen werden:

````python
x = 3
y = 0
comparison = x > y  # Zuweisung
print(comparison)
````

Die Variable `comparison` speichert das Ergebnis des Vergleichs und kann an anderer Stelle im Programm verwendet werden. Es folgt ein Beispiel, welches auf die `if`-`else`-Verzweigung vorgreift (Sie können es sicher trotzdem schon verstehen): 

````python
if comparison:
    print(x, 'ist größer als', y)
else:
    print(x, 'ist kleiner oder gleich', y)
````

````python
# Zur Erinnerung: Wir legen unsere
# Variablennamen selbst fest. Hier zur Demo dasselbe
# Beispiel erneut mit einem Quatsch-Namen:
x = 3
y = 0
dagobert_duck = x > y  # Zuweisung
print(dagobert_duck)
# Wir könnten jetzt mit dagobert_duck
# alles tun, was wir zuvor mit comparison
# getan haben.
````

Wahrheitswerte tauchen nicht nur bei Vergleichen von Zahlen auf, sondern auch in vielen anderen Situationen, wie zum Beispiel bei der Suche nach einem Buchstaben in einem Wort. Das würde in Python so aussehen:

````python
# Suche nach einem 'a' in 'apple':
'a' in 'apple' 
````

````python
# Suche nach einem 'a' in 'lemon'
'a' in 'lemon' 
````

````python
# Achtung: Kleinbuchstaben != Großbuchstaben
'a' in 'YOGA'
````

Mit der Negation `not` kehren wir einen Wahrheitswert in sein Gegenteil um, wie die folgenden Beispiele illustrieren:

````python
not (4 < 3)
````

````python
not ('a' in 'apple')  # idiomatischer und gleichwertig: 'a' not in 'apple'
````

Einfache logische Ausdrücke können durch die Verknüpfungen `or` und `and` zu komplexeren Bedingungen kombiniert werden. Wenn zwei Bedingungen mit `and` verknüpft werden, ist die kombinierte Bedingung genau dann wahr, wenn **beide** Einzelbedingungen wahr sind: 

````python
5 > 4 and 4 > 3 
````

````python
5 > 4 and 4 > 6
````

Verknüpft man hingegen mit `or`, so genügt es, dass eine der beiden Einzelbedingungen wahr ist: 

````python
5 > 4 or 4 > 6 # mindestens eine Bedingung wahr.
````

````python
5 > 4 or 4 > 3 # beide wahr
````

Probieren Sie eigene Beispiele mit `and`, `or` und `not` aus!

### Comparison Chains in Python {.unnumbered}

Es gibt in Python eine Möglichkeit, mehrere Vergleiche miteinander zu verketten, wie man es z.B. aus der Mathematik kennt. Die Vergleiche in der Kette werden einzeln ausgewertet und mit `and` verknüpft: 

|Comparison Chain | Bedeutung|
|---|---|
|`18.5 <= x < 25` | `18.5 <= x and x < 25` |
|`x < y > z ` | `x < y and y > z`  |
|`x == y == z`| `x == y and y == z` |
|`x != y != z` | `x != y and y != z` |

Während das erste Beispiel genauso aussieht, wie man es aus der Mathematik kennt, sieht das zweite Beispiel vielleicht etwas seltsam aus. Die Vergleiche werden aber einfach der Reihe nach abgearbeitet, unabhängig davon, welche Vergleichsoperatoren verwendet werden. Das dritte Beispiel prüft, ob alle Variablen gleich sind. Trügerisch ist das vierte Beispiel `x != y != z`, da es **nicht** feststellt, ob alle Variablen ungleich sind (`x` und `z` werden nicht miteinander verglichen!).

**Tipp**: Setzen Sie Comparison Chains mit Bedacht ein. Am intuitivsten sind sie, wenn die Ausdrücke so aussehen, wie man es aus der Mathematik kennt (vgl. erste Zeile in der Tabelle). 
:::

## Zum Abschluss des Kapitels: ein wenig Fachsprache 

Wir können bereits eine Menge tun in Python. An dieser Stelle wollen wir kurz auf das bisher Gelernte zurückschauen und dabei wichtige Fachtermini einführen, die Ihnen immer wieder begegnen werden. Es handelt sich um die Begriffe "Anweisung", "Ausdruck" und "Literal".

::: {.callout-note appearance="minimal"}
**Anweisung (engl. statement):** 
Ein vollständiger Befehl, der ausgeführt wird – meist eine Zeile.
:::

::: {.callout-note appearance="minimal"}
**Ausdruck (engl. expression):** 
Code, der ausgewertet wird und einen Wert liefert, z.B. `5` (liefert 5), `x` (liefert den Wert von x), `x + 5` (liefert den Wert von x + 5), `x > y` (liefert True oder False) oder `len("Apfel")` (liefert 5).
:::

Wir haben bereits verschiedene Arten von Ausdrücken kennengelernt: **arithmetische** und **logische Ausdrücke**.
Arithmetische Ausdrücke rechnen mit Zahlen und liefern eine Zahl als Ergebnis. Logische Ausdrücke (auch *boolesche Ausdrücke* genannt) liefern einen Wahrheitswert und arbeiten typischerweise mit Vergleichen sowie `and`, `or` und `not`. Zuweisungen sind selbst keine Ausdrücke - sondern Anweisungen - aber sie erwarten auf ihrer rechten Seite einen Ausdruck.  

````python
#| output: false
5 + 3  # Ein arithmetischer Ausdruck. Wert: 8
x = 5 + 3 # Eine Zuweisung ist eine Anweisung; die rechte Seite ist ein Ausdruck.
3 * x + 4  # Ein arithmetischer Ausdruck. Wert: 28
y = 3 * x + 4 # Eine Zuweisung ist eine Anweisung; die rechte Seite ist ein Ausdruck. 
````

In manchen Programmiersprachen sind dagegen auch Zuweisungen Ausdrücke: In C weist `x = 3` der Variablen `x` den Wert 3 zu und liefert gleichzeitig den Wert 3. Deshalb ist es in C nicht überraschend, dass `y = x = 3` funktioniert: Zuerst wird `x = 3` ausgeführt, und der gelieferte Wert 3 wird anschließend `y` zugewiesen.

In Python funktioniert `y = x = 3` ebenfalls, wird intern aber anders verarbeitet: Der Interpreter erkennt eine Zuweisung mit zwei Zielen, wertet den Ausdruck `3` einmal aus und weist ihn von links nach rechts zu, also zuerst `y = 3` und dann `x = 3`.

````python
#| output: false
4 > 5  # Ein logischer Ausdruck. Wert: False.
1 == 1 and 5 > 4 # Ein logischer Ausdruck. Wert: True.
comparison = 99 > 42  # Eine Zuweisung ist eine Anweisung; die rechte Seite ist ein Ausdruck.
````

Weiterhin stellen Funktionsaufrufe einen Ausdruck dar, wie zum Beispiel:

````python
#| output: false
len('Apfel')  # len() liefert die Länge, hier die Anzahl der Zeichen: 5
x = len('Apfel') # der Ausdruck wird als rechte Seite einer Zuweisung verwendet
````

Auch Variablen selbst können als Ausdruck betrachtet werden, denn sie geben ihren Wert zurück:

````python
#| output: false
z = 100
z  # Ein Ausdruck; liefert den Wert von z, also 100.
y = z # Eine Zuweisung mit dem Ausdruck z auf der rechten Seite.
````

::: {.callout-note appearance="minimal"}
**Literal (vom lateinischen Wort für Buchstabe):** 
Ein Literal ist ein fester, unveränderlicher Wert (wie eine Zahl, eine Zeichenkette oder ein Wahrheitswert), der direkt im Code verwendet wird. Da Literale ausgewertet werden können, um einen Wert zu liefern, gelten sie ebenfalls als Ausdrücke.
:::

Es folgen einige Beispiele (die Datentypen erklären wir später):

````python
#| output: false
"Hello, World"  # String-Literal (Zeichenkette)
4.2  # Float-Literal (Gleitkommazahl)
42  # Integer-Literal (Ganzzahl)
True  # Boolean-Literal (Wahrheitswert)

# Alle diese Literale sind gleichzeitig Ausdrücke:
# Sie geben sich selbst als Wert zurück.
````
