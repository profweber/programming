## Prozedurale Programmierung

Bisher haben wir ausschließlich in der sog. **imperativen Programmierung** gearbeitet: Dabei wird der Programmablauf durch eine Abfolge von Befehlen gesteuert, die der Reihe nach abgearbeitet werden (der Kontrollfluss wird insbesondere durch Verzweigungen und Schleifen beeinflusst).

Die **prozedurale Programmierung** ist ein Ansatz innerhalb der imperativen Programmierung, bei dem das Programm als eine Reihe von Prozeduren oder Funktionen organisiert wird. Durch diese Aufteilung können komplexe Abläufe in einzelne, handhabbare Teile zerlegt werden, was zu einem klareren und besser wartbaren Code führt.

### Was ist eine Funktion? {.unnumbered}
Eine Funktion ist ein in sich geschlossener Block von Anweisungen, der eine bestimmte Aufgabe erfüllt. Vergleichen Sie eine Funktion mit einem Kochrezept: Das Rezept beschreibt eine schrittweise Anleitung zur Zubereitung eines Gerichts. Wann immer Sie das Gericht kochen, verwenden Sie dasselbe Rezept wieder und wieder. In der Programmierung würden wir sagen: "Sie rufen dieselbe Funktion immer wieder auf." Von dem Rezept benötigen Sie aber nur ein einziges Exemplar (ganz egal, wie oft Sie das Gericht kochen werden). So ist es auch mit dem Code, der in einer Funktion definiert wird: Sie schreiben ihn ein einziges Mal und verwenden ihn dann beliebig oft. 

Manchmal wird auch eine Analogie zu mathematischen Funktionen $f(x)$ gezogen. Diese Analogie hat aber einige Schwächen: Die mathematische Funktion bekommt immer einen *Input* (hier $x$) und gibt immer einen Wert als *Output* ($y = f(x)$) zurück. Das muss bei Funktionen in der Programmierung nicht sein! Eine Funktion in der Programmierung kommt auch ohne Input und Output aus, wie wir gleich sehen werden. Es liegt ganz in Ihrer Freiheit, ob Ihre Funktionen mit Inputs und Outputs arbeiten oder nicht. Sie können sogar eine Funktion programmieren, die beliebig viele Inputs verarbeiten kann (eine solche Funktion haben wir bereits kennengelernt, wissen Sie, welche?).      

Bevor wir konkret werden, fassen wir die Vorteile von Funktionen zusammen:

* **Modularisierung:** Zerlegung des Codes in kleine, verständliche Einheiten.

* **Wiederverwendbarkeit:** Einmal definierte Funktionen können mehrfach eingesetzt werden.

* **Lesbarkeit und Wartbarkeit:** Klar strukturierter Code erleichtert das Verständnis und die Fehlersuche.

* **Vermeidung von Redundanz:** Gleicher Code muss nicht mehrfach geschrieben werden, sondern kann über Funktionsaufrufe wiederverwendet werden. Insbesondere von Vorteil, wenn Sie etwas ändern müssen: Sie müssen nur eine Stelle korrigieren, egal wie oft die Funktion aufgerufen wird.

## Built-in Functions

In Python gibt es eine Vielzahl von *Built-in Functions* (etwa 70 Funktionen, je nach Version). Das sind Funktionen, die ohne zusätzliche Importe direkt verwendet werden können. Viele davon haben wir bereits kennengelernt und wollen sie in diesem Abschnitt einmal übersichtlich darstellen (außerdem ergänzen wir ein paar nützliche Funktionen, die wir noch nicht behandelt haben):

### Funktionen für Ein- und Ausgabe {.unnumbered}
`print(*args)` – erzeugt eine Konsolenausgabe (das Symbol `*` zeigt an, dass beliebig viele Argumente erlaubt sind, wie wir später noch sehen werden).

`input(prompt)` – zeigt eine Aufforderung (`prompt`) an und liest die Benutzereingabe als **String** ein.

### Funktionen für Typumwandlungen (Casting) {.unnumbered}
`int(x)`, `float(x)`, `str(x)`, ... – konvertiert in den entsprechenden Datentyp. 

Diese Funktionen können auch genutzt werden, um leere Instanzen zu erzeugen, z.B. `str()` (ein String ohne Zeichen) oder `list()` (eine Liste ohne Elemente).

### Allgemeine Informationsfunktionen {.unnumbered}
`type(x)` – gibt den Datentyp einer Instanz zurück.

`id(x)` – liefert die eindeutige Identität einer Instanz.

`help(x)` – zeigt die Hilfeseite an (kann mit Objekten oder Funktionen verwendet werden). 

````python
# Beispiele zur Verwendung von help:
print("Hilfe für die Funktion len:")
help(len)
print("\nHilfe für die Funktion bin:")
help(bin)
````

### Numerische Funktionen {.unnumbered}
`abs(x)` – Betrag einer Zahl (funktioniert auch mit `complex`).

`round(x, n)` – rundet eine Zahl auf `n` Nachkommastellen. Ohne `n` entsteht ein `int`. Achtung: Genau in der Mitte rundet Python zur geraden Zahl, `round(2.5)` ergibt `2` (siehe Kapitel *Datentypen*).

````python
# Beispiele zur Verwendung von abs und round:
print(abs(-7.8234))
print(abs(1+1j))  # Betrag einer komplexen Zahl
print(round(-7.8234, 2))
print(round(-7.8234, 1))
print(round(-7.8234)) # ohne Angabe von n
````

### Nützliche Funktionen für Sequenzen {.unnumbered}
`range(start, stop, step)` – erzeugt eine Zahlenfolge.

`len(x)` – gibt die Länge einer Sequenz zurück, z.B. Anzahl der Zeichen von einem String oder Anzahl der Elemente in einer Liste.

`sum(x)` – summiert alle Werte in einer Sequenz.

`max(x)` und `min(x)` – gibt das Maximum bzw. Minimum in einer Sequenz zurück.

`sorted(x)` – erzeugt eine sortierte Liste mit den Elementen der Sequenz und gibt diese zurück. (Hinweise: Das Original bleibt unverändert! Auch wenn man `sorted()` auf ein `set` oder `tuple` anwendet, erhält man eine Liste.)

````python
# Beispiele zur Verwendung von len, sum, max, min und sorted:
my_list = [0.5, 1, 0.25, -0.5, 2]
print("len:", len(my_list))
print("sum:", sum(my_list))
print("max:", max(my_list))
print("min:", min(my_list))
print("sorted:", sorted(my_list))
````

### Funktionen für verschiedene Zahlensysteme {.unnumbered}
`bin(x)` – wandelt eine ganze Zahl in eine Binärdarstellung (z.B. `bin(10) → '0b1010'`)

`oct(x)` – wandelt eine ganze Zahl in eine Oktaldarstellung (z.B. `oct(10) → '0o12'`)

`hex(x)` – wandelt eine ganze Zahl in eine Hexadezimaldarstellung (z.B. `hex(255) → '0xff'`)

Hinweis: Diese Funktionen liefern **Strings** zurück.

### Funktionen für Zeichencodierung {.unnumbered}
`ord(char)` – gibt den Unicode-Codepunkt eines Zeichens zurück, z.B. `ord('A') → 65`, `ord('ß') → 223`.

`chr(code)` – gibt das Zeichen zum angegebenen Codepunkt zurück, z.B. `chr(65) → 'A'`, `chr(223) → 'ß'`

## Eigene Funktionen programmieren
Python bietet eine einfache und flexible Syntax zur Definition von Funktionen. Eine Funktion wird mit dem Schlüsselwort `def` eingeleitet, gefolgt vom Funktionsnamen (den Sie selbst festlegen). In runden Klammern werden optional Eingangsparameter definiert, dann folgt ein Doppelpunkt. Der eigentliche Funktionsblock muss eingerückt werden (das kennen Sie schon von Code-Blöcken in Verzweigungen und Schleifen). Hier ein einfaches Beispiel ohne Eingangsparameter:  

````python
def greeting():
    print("Hello, world!")
````

Wir haben eine sehr primitive Funktion definiert, die nichts anderes tut, als `"Hello, world!"` auszugeben, wann immer wir sie aufrufen:

````python
greeting()
````

Beachten Sie beim Aufruf der Funktion die runden Klammern, die gleich mehr Sinn ergeben werden. 

### Parameter und Argumente {.unnumbered}
Wie bereits erwähnt, können Funktionen **Parameter** entgegennehmen, um mit variablen Eingabewerten zu arbeiten. Zum Beispiel können wir unsere Begrüßungsfunktion so erweitern, dass sie einen Namen als Parameter entgegennimmt und diesen für eine personalisierte Begrüßung verwendet: 

````python
def greeting(name):
    print("Hallo,",name)
````

````python
greeting("Anna")
````

````python
greeting("Tom")
````

::: {.callout-note}
### Argumente {.unnumbered}
Den konkreten Wert, den man an eine Funktion übergibt, nennt man **Argument**. Wir haben in unseren Beispielen bereits verschiedene Argumente ("Anna","Tom") für den Parameter `name` verwendet.
:::

Was jetzt nicht mehr funktioniert: Wir können die Funktion nicht mehr ohne ein Argument für den Parameter `name` aufrufen! Probieren Sie `greeting()` und Python wird sich beschweren, dass ein Argument fehlt! Eine besondere Möglichkeit, dieses Problem zu beheben, ist es, sog. **Standardargumente** für Parameter festzulegen. Diese werden verwendet, wenn der Anwender der Funktion kein Argument übergibt (ob das sinnvoll ist oder nicht, hängt ganz von Ihrer Anwendung ab). Wir demonstrieren die Begrüßungsfunktion mit einem Standardargument:  

````python
def greeting(name = "Welt"):
    print("Hallo,",name)
````

````python
greeting("Lisa")
````

````python
greeting()
````

Der Parameter `name` wurde durch die Definition eines Standardarguments zu einem **optionalen Parameter**: Es ist nicht mehr zwingend erforderlich, ihn beim Funktionsaufruf festzulegen.

### Mehrere Parameter {.unnumbered}

Selbstverständlich können wir mehr als einen Parameter verwenden: 

````python
def greeting(name, age):
    print("Hallo,", name)
    if age < 18:
        print("Sie sind noch minderjährig!")
    else:
        print("Willkommen im Club!")
````

````python
greeting("Tom", 16)
greeting("Lisa", 21)
````

Jetzt muss man allerdings aufpassen, wenn man Standardargumente verwenden will, denn es gibt eine wichtige Regel:

:::{.callout-caution}
### Optionale Parameter immer hinter nicht-optionalen Parametern! {.unnumbered}
Folgende Definition ist korrekt: 

```python
def greeting(name, age = 20):
    # Funktionsblock
    ...
```

Folgender Code ist **falsch** und wird nicht funktionieren:

```python
def greeting(name="Peter", age):
    # Funktionsblock
    ...
```
:::

### Rückgabewerte {.unnumbered}

Funktionen können einen Wert zurückgeben. Hierfür verwendet man (wie in C oder C++) das Keyword `return`:

````python
def square(x):
    return x * x

result = square(4)
print(result)
````

Wir können die Funktion wieder beliebig oft mit verschiedenen Argumenten aufrufen: 

````python
print(square(3))
print(square(-2))
print(square(7.3))
````

Und was gibt eine Funktion ohne `return` zurück, z.B. unsere Funktion `greeting` von oben? Den speziellen Wert `None` („nichts“). Daraus ergibt sich ein typischer Anfängerfehler: Man gibt ein Ergebnis in der Funktion mit `print` aus, statt es mit `return` zurückzugeben. Das folgende Beispiel zeigt **absichtlich, wie man es nicht machen sollte**:

````python
# Negativbeispiel: print statt return!
def square_print(x):
    print(x * x)  # gibt das Ergebnis nur auf dem Bildschirm aus

result = square_print(4)
print("Rückgabewert:", result)
````

Die Funktion zeigt zwar `16` an, ihr Rückgabewert ist aber `None` – mit dem Ergebnis kann man also nicht weiterrechnen. Richtig ist die Variante `square` von oben mit `return`.

::: {.callout-important}
### Merke: `return` beendet sofort die Funktion! {.unnumbered}
Innerhalb einer Funktion können wir den Kontrollfluss mit allen uns bekannten Mitteln beeinflussen, z.B. mit Verzweigungen. Wichtig ist aber, dass die Funktion sofort verlassen wird, wenn der Interpreter auf ein `return` trifft!
:::

Eine Besonderheit in Python ist, dass wir sogar mehr als einen Wert zurückgeben können! Das geht in vielen anderen Programmiersprachen nicht ohne Weiteres, in C oder C++ würde man dafür z.B. Datenstrukturen (als Container für mehrere Werte) oder Zeiger verwenden (die auf einen Bereich im Speicher verweisen, wo die Werte hintereinander abgelegt werden). In Python geht das ganz einfach, wie das folgende Beispiel zeigt:   

````python
def int_division(a, b): 
    # Division mit Rest
    # zwei Rückgabewerte durch Komma getrennt:
    return a // b, a % b 

a = 13
b = 3

quotient, remainder = int_division(a, b)

print(a, "geteilt durch", b ,"ergibt:")
print(quotient, "mit Rest", remainder)
````

Genau genommen gibt die Funktion dabei einen einzigen Wert zurück, nämlich das Tupel `(4, 1)`, das beim Aufruf in die beiden Variablen *entpackt* wird (mehr zu Tupeln im Kapitel *Datenstrukturen*). Die Reihenfolge der Rückgabewerte ist natürlich zu beachten, sonst geschehen leicht Fehler (würde man versehentlich `remainder, quotient = int_division(a, b)` schreiben, stünde in `quotient` der Rest und in `remainder` der Quotient). 

Übrigens gibt es genau diese Funktion in Python bereits eingebaut: `divmod(13, 3)` liefert `(4, 1)`. Unsere eigene Funktion haben wir bewusst anders genannt, denn eine Funktion namens `divmod` würde die eingebaute Funktion überdecken. 

### Keyword-Argumente {.unnumbered}

Python erlaubt uns, Funktionen mit sog. **Keyword-Argumenten** aufzurufen. Damit ist gemeint, dass die Argumente explizit benannt werden, anstatt sie anhand ihrer Reihenfolge zu identifizieren. Betrachten wir zunächst folgendes Beispiel, in dem wir eine Funktion wie gewohnt aufrufen:

````python
def print_personal_details(name, age, city):
    print("Name:", name)
    print("Alter:", age)
    print("Wohnort:", city)

print_personal_details("Alice", 23, "London") # wie gehabt
````

Jetzt zeigen wir Varianten mit Keyword-Argumenten:

````python
# mit Verwendung von Keyword-Argumenten:
print_personal_details(name = "John", city = "Birmingham", age = 32) 
````

````python
# teilweise mit Keyword-Argumenten:
print_personal_details("Marc", city = "Hamburg", age = 19)
````

Wie Sie sehen, können Sie die Parameternamen explizit verwenden, um ein Argument zu definieren, und dabei sogar von der Reihenfolge im Funktionskopf abweichen. Eine Regel gibt es allerdings auch hier wieder: Wenn Sie nur teilweise Keyword-Argumente verwenden (wie im Beispiel `"Marc"` oben), kommen die Keyword-Argumente immer am Schluss! Zur namentlichen Abgrenzung von den Keyword-Argumenten bezeichnet man die Argumente, die anhand ihrer Reihenfolge identifiziert werden, als **Positionsargumente** (engl.: *positional arguments*). Halten wir also fest: 

:::{.callout-caution}
### Positionsargumente zuerst, dann Keyword-Argumente! {.unnumbered}
Python weist zunächst die Argumente entsprechend ihrer Reihenfolge den Parametern zu (Positionsargumente).
Danach können verbleibende Parameter explizit über Schlüsselwörter (Keyword-Argumente) gesetzt werden.

## Dynamische Typisierung und mögliche Gefahren

Wir kennen bereits die dynamische Typisierung als besondere Eigenschaft von Python. Im Zusammenhang mit Funktionen erlaubt sie uns, dieselbe Funktion mit unterschiedlichen Datentypen aufzurufen (wir haben das bereits oben einmal getan, finden Sie das Beispiel?). In anderen Programmiersprachen muss man dafür die Funktion in der Regel mehrfach implementieren (eine Variante für jeden Datentyp), man sagt auch: Die Funktion wird **überladen**. Bei überladenen Funktionen in C++ (in C gibt es kein Überladen) erkennt der Compiler anhand der übergebenen Datentypen, welche Version der Funktion gemeint ist. 

Nicht so in Python, wo der Datentyp der Parameter und Rückgabewerte nicht vom Programmierer bzw. der Programmiererin festgelegt wird. Das kann sehr nützlich sein, aber auch zu Fehlern führen, wenn die Funktion mit inkompatiblen Datentypen aufgerufen wird. Hierzu ein Beispiel: 


````python
def multiply(x, y):
    return x * y
````

Eine sehr einfache Funktion, die zwei Parameter miteinander multipliziert. Wir rufen sie jetzt mit unterschiedlichen Datentypen auf: 

````python
print(multiply(3, 4)) # zwei Integer
print(multiply(2, 3.2)) # Integer und Float
print(multiply(2+2j, 3-1j)) # zwei komplexe Zahlen
print(multiply("hi", 3)) # String und Integer
````

Hätten Sie geahnt, was alles mit dieser kleinen, unscheinbaren Funktion möglich ist?

Es scheint fast alles möglich zu sein, aber eben nur *fast*! Rufen Sie die Funktion mit zwei Strings auf und Ihr Programm stürzt ab (wie sollen zwei Strings miteinander multipliziert werden?). Natürlich sind auch viele weitere Kombinationen nicht möglich (versuchen Sie z.B. zwei Listen miteinander zu multiplizieren oder eine Liste mit einem String). Manchmal ist es sogar noch schlimmer, wenn das Programm zwar nicht abstürzt, aber das Ergebnis völlig unerwartet ist (hätten Sie z.B. das Ergebnis der Multiplikation von String mit Integer in der Form erwartet?). Dann entstehen an einer anderen Stelle im Code womöglich überraschend Probleme und Sie müssen langwierig debuggen, bis Sie die Ursache finden.    

Daran erkennen Sie wieder: Python bietet große Flexibilität, aber Sie müssen damit verantwortungsbewusst umgehen!

### Ergänzung: Mit `isinstance` den Typ von Parametern prüfen {.unnumbered}

Mit der Funktion `isinstance` kann bei Bedarf der Typ von Übergabeparametern geprüft werden. Es ist somit möglich, eine Funktion gegen die Verwendung mit inkompatiblen Typen abzusichern. Im Beispiel sieht dies so aus: 

````python
def add_and_sort_lists(a, b):
    if isinstance(a, list) and isinstance(b, list):
        c = a + b
        c.sort()
        return c
    else:
        print("Inkompatible Typen")
````

````python
add_and_sort_lists([1, 5, 2], [4, 3])
````

````python
add_and_sort_lists([1, 2], 3.4)
````

Hinweis: Eine Abfrage mit `type()` (z.B. `type(x) == int`) wird für diesen Zweck nicht empfohlen, da `isinstance()` im Gegensatz zu `type()` auch abgeleitete Datentypen erkennt (siehe Vererbung / objektorientierte Programmierung: Wenn eine Klasse `Dog` von einer Klasse `Animal` erbt, würde `isinstance` einen `Dog` auch als `Animal` erkennen). 

## Funktionen mit beliebig vielen Parametern

Wir kennen seit Beginn des Kurses eine Funktion, die beliebig viele Argumente entgegennehmen kann:

````python
print()  # 0 Argumente
print("Hallo")  # 1 Argument
print("Hallo,", "Welt")  # 2 Argumente usw.
````

Mit einem einfachen Trick können Sie selbst solche Funktionen definieren: Stellen Sie vor einen Parameter ein `*`. Dann können beliebig viele Argumente übergeben werden; innerhalb der Funktion stehen sie als Tupel zur Verfügung, über das Sie z.B. mit der bekannten for-Schleife iterieren können:

````python
def sum_numbers(*numbers):
    result = 0
    
    for number in numbers:
        result += number

    return result

print(sum_numbers(-2))
print(sum_numbers(5, 10))
print(sum_numbers(1, 2, 3, 4)) 
````

## Scope

In Python – und generell in der Programmierung – bezieht sich der Scope (Gültigkeitsbereich) auf den Bereich des Codes, in dem eine Variable sichtbar und zugänglich ist. Es gibt dabei zwei Hauptarten von Scope:

* **Lokaler Scope**: Eine Variable, die innerhalb einer Funktion definiert wird, ist nur in dieser Funktion sichtbar und außerhalb nicht zugänglich.

* **Globaler Scope:** Eine Variable, die außerhalb einer Funktion definiert wird, ist nach ihrer Definition im gesamten Programm sichtbar (auch innerhalb einer Funktion). Python schützt uns aber davor, globale Variablen unbeabsichtigt im Inneren einer Funktion neu zuzuweisen, es sei denn, wir machen unsere Absicht explizit deutlich. Dafür gibt es in Python das Keyword `global`.

Beginnen wir mit einem Beispiel zum lokalen Scope: 

````python
def my_function(a, b):
    c = a + b
    return c**2

my_function(2,3)
print(c) # FEHLER!!!! c ist hier nicht bekannt. 
````

Auf die Variable `c` kann außerhalb der Funktion nicht zugegriffen werden, sie existiert nur im lokalen Scope der Funktion.  

Nun versuchen wir umgekehrt innerhalb der Funktion auf eine Variable zuzugreifen, die außerhalb definiert wurde: 

````python
def greeting():
    print("Hallo,", name)  # name existiert im globalen Scope

name = "Peter"
greeting()
````

Das Beispiel ist geglückt! Allerdings handelt es sich hierbei nicht um guten Programmierstil. Vielmehr sollte man `name` als Parameter der Funktion definieren. Was geschieht eigentlich, wenn es `name` im globalen und im lokalen Scope gibt?

````python
def greeting(name):
    print("Hallo,", name)

name = "Peter"
greeting("Tina")
````

**Der lokale Scope geht vor! Die Funktion sucht immer zuerst in ihrem lokalen Scope nach der Variable. Nur wenn sie sie dort nicht findet, schaut sie im globalen Scope nach.** Diese Regel nennt man auch LEGB-Rule (@fig-scope-legb): Python sucht nach Variablennamen *von innen nach außen* über diese Stufen: L = local scope, E = enclosing scope (umschließende Funktionen), G = global scope, B = built-in scope (eingebaute Python-Bezeichner). 

![Python sucht einen Namen von innen nach außen: Local → Enclosing → Global → Built-in. Im Beispiel wird der lokale Parameter `name` (Wert `"Tina"`) zuerst gefunden (einen Enclosing Scope gibt es hier nicht).](images/scope_legb.png){#fig-scope-legb}

### Das Keyword `global` in Python {.unnumbered}

Wie bereits erwähnt, schützt uns Python davor, unbeabsichtigt eine Variable aus dem globalen Scope neu zuzuweisen, wenn wir uns innerhalb einer Funktion befinden. Das wird im folgenden Beispiel illustriert:

````python
def change_name(new_name):
    name = new_name

name = "Peter"
change_name("John")
print(name)
````

Hätten Sie die Ausgabe "John" erwartet? Python schützt uns: Beim Versuch, die globale Variable innerhalb einer Funktion neu zuzuweisen, hat Python stattdessen eine neue Variable im lokalen Scope der Funktion angelegt - mit demselben Bezeichner: `name`! Diese lokale Variable bekam den Wert "John", aber die Variable im globalen Scope wurde nicht angefasst. Machen wir das ruhig explizit:  

````python
def change_name(new_name):
    name = new_name
    print("name innerhalb der Funktion:", name)

name = "Peter"
change_name("John")
print("name außerhalb:", name)
````

Wir haben nun bereits mehrfach gesehen, dass Variablen mit demselben Namen in unterschiedlichen Scopes existieren können. Python speichert die Variablennamen für jeden Scope in einer eigenen Zuordnungstabelle (Name → Instanz), dem sog. Namespace, der zu diesem Scope gehört. Im obigen Beispiel existiert also eine Variable `name` im globalen Namespace und eine Variable `name` im Namespace der Funktion `change_name`. Obwohl diese beiden Variablen gleich heißen, handelt es sich doch um getrennte Variablen, die in der Regel unterschiedliche Instanzen referenzieren. Beim Aufruf von `print` innerhalb der Funktion wird gemäß der LEGB-Rule die lokale Variable zuerst gefunden und verwendet. Bei der Verwendung von `print` außerhalb der Funktion findet Python stattdessen die globale Variable (Reminder: Python sucht immer *von innen nach außen*).

Eine Feinheit führt immer wieder zu Verwirrung: Ob ein Name in einer Funktion lokal ist, entscheidet Python nicht erst, wenn die Zeile ausgeführt wird, sondern bereits beim Einlesen der Funktion. Sobald einem Namen **irgendwo** in der Funktion ein Wert zugewiesen wird, gilt er in der **gesamten** Funktion als lokal – auch in den Zeilen davor:

````python
counter = 0

def increment():
    counter += 1  # FEHLER! counter gilt hier als lokale Variable ohne Wert

increment()
````

`counter += 1` ist eine Zuweisung an `counter`. Deshalb behandelt Python `counter` in der ganzen Funktion als lokale Variable – und die hat beim Lesen noch keinen Wert. Abhilfe schafft das Keyword `global` (siehe gleich) oder, besser, ein Parameter und ein Rückgabewert.

Zurück zu unserem Beispiel `change_name`: Den vorgestellten Sicherheitsmechanismus kann man mit dem Keyword `global` bewusst aushebeln!

````python
def change_name(new_name):
    global name  # hey Python, wir meinen die globale Variable!
    name = new_name

name = "Peter"
change_name("John")
print(name)
````

Mit der Zeile `global name` weisen wir Python explizit an, die Variable `name` aus dem globalen Scope zu verwenden! In diesem Beispiel wird **keine** lokale Variante von `name` angelegt. Die Variable `name` existiert nur im globalen Namespace, kann aber in der Funktion frei verwendet werden.  

::: {.callout-tip}
### Verwendung globaler Variablen in Funktionen {.unnumbered}
Finden Sie das alles noch ein wenig verwirrend? Gut so, denn Sie sollten von der Verwendung globaler Variablen in Funktionen sowieso besser absehen. Versuchen Sie stattdessen saubere Schnittstellen mithilfe von Parametern und Rückgabewerten zu definieren. Wann immer Sie geneigt sind, lesend auf einen globalen Wert zuzugreifen, übergeben Sie diesen lieber als Argument. Wann immer Sie geneigt sind, schreibend auf einen globalen Wert zuzugreifen, arbeiten Sie lieber mit einem Rückgabewert, den Sie im Hauptprogramm zuweisen können.

Beispiel: Eine Zahl im globalen Scope soll durch ihre Quadratzahl ersetzt werden. 
```python
def square(x):
    return x**2 # die Funktion führt die Rechnung durch

a = 42
a = square(a) # die eigentliche Veränderung geschieht außerhalb der Funktion
```
:::

## Funktionsaufrufe mit mutable Datentypen

Wir haben bereits den Unterschied zwischen mutable und immutable Datentypen kennengelernt, insbesondere das Phänomen von Seiteneffekten, wenn mehrere Referenzen auf eine veränderliche Instanz zeigen.
Seiteneffekte können auch durch den Aufruf von Funktionen auftreten: Wir übergeben an Funktionen eine Referenz auf eine Instanz. Wenn die Instanz mutable ist, kann die Funktion die Instanz verändern! Hierzu eine Demo:  

````python
def append_a(l, a):
    l.append(a)

my_list = [1, 2, 3]
append_a(my_list, 100)

print(my_list)  # my_list wurde verändert
````

In diesem Beispiel zeigen innerhalb der Funktion `append_a` zwei Referenzen auf dieselbe mutable Instanz: einmal die Variable `my_list` aus dem globalen Namespace und einmal die Variable `l` im lokalen Namespace der Funktion. Die Veränderung, die `l` durch den Aufruf von `append` bewirkt, sieht deshalb auch die Variable `my_list`. Vorsicht beim Operator `+=`: Schreibt man in der Funktion `l += [a]`, wird die Liste des Aufrufers ebenfalls verändert (wie bei `append`). Mit `l = l + [a]` entsteht dagegen eine neue Liste, die nur innerhalb der Funktion bekannt ist – `my_list` bleibt dann unverändert (mehr dazu im Kapitel *Datenstrukturen*).  

Besonders überraschend ist vielleicht das folgende Beispiel: 

````python
def append_a_to_global_list(a):
    my_list.append(a) # my_list wird über die LEGB-Rule im globalen Namespace gefunden

my_list = [1, 2, 3]
append_a_to_global_list(100)

print(my_list)  # my_list wurde verändert
````

Wundern Sie sich, dass wir das Keyword `global` nicht gebraucht haben? 
Der Grund ist einfach: In der Funktion wurde keine neue Zuweisung an `my_list` vorgenommen. Stattdessen haben wir lediglich die bereits existierende Instanz verändert – und das funktioniert nur bei mutable Datentypen!
Das Keyword `global` wäre nur dann erforderlich gewesen, wenn wir `my_list` innerhalb der Funktion eine neue Instanz zugewiesen hätten, also z.B. mit `my_list = [4, 5, 6]`. In diesem Fall müssten wir – wie bekannt – mit `global my_list` deutlich machen, dass sich die Zuweisung auf die globale Variable beziehen soll (und nicht nur auf eine lokale im Namespace der Funktion). Achtung: Auch `my_list += [4]` zählt als Zuweisung und führt ohne `global` zum `UnboundLocalError` (siehe oben). 

Es sei darauf hingewiesen, dass die Funktion aus dem letzten Beispiel schlechter Programmierstil ist: Viel verständlicher ist es, wenn die Liste, die verändert werden soll, als Parameter übergeben wird und nicht erst über die LEGB-Rule gefunden werden muss. Außerdem funktioniert der Code nicht mehr, wenn die Variable im globalen Namespace umbenannt wird und die Namensänderung innerhalb der Funktion vergessen wird.

::: {.callout-warning}
### Fallstrick: Eine Liste als Standardwert eines Parameters {.unnumbered}
```python
def add_measurement(value, measurements=[]):
    measurements.append(value)
    return measurements

print(add_measurement(4.8))   # [4.8]
print(add_measurement(5.1))   # [4.8, 5.1] – der erste Wert ist noch da!
```

Der Standardwert `[]` wird nur **ein einziges Mal** erzeugt, nämlich wenn Python die `def`-Zeile ausführt – nicht bei jedem Aufruf neu. Alle Aufrufe ohne zweites Argument teilen sich deshalb dieselbe Liste, und weil Listen mutable sind, sammeln sich die Werte darin an. Richtig ist es, `None` als Standardwert zu verwenden und die Liste erst in der Funktion zu erzeugen:

```python
def add_measurement(value, measurements=None):
    if measurements is None:
        measurements = []     # bei jedem Aufruf eine neue Liste
    measurements.append(value)
    return measurements
```
:::

## Docstrings und Typ-Annotationen: Meine Funktion für andere dokumentieren
In großen Software-Projekten sind Sie nicht die einzige Person, die Ihre Funktionen benutzt. Sie können durch verständliche Dokumentation die Nutzbarkeit Ihrer Funktionen enorm erhöhen. Eine einfache Möglichkeit sind sog. Docstrings: Dabei handelt es sich um eine spezielle Kommentierung, direkt nach der definierenden Zeile:

````python
def bending_stress(moment, y, inertia):
    """
    Berechnet die Biegespannung in einem Balken.

    Formel: sigma = (M * y) / I

    Args:
        moment (float): Biegemoment in Newton-Meter (Nm).
        y (float): Abstand von der Neutralachse in Meter (m).
        inertia (float): Flächenträgheitsmoment in m^4.

    Returns:
        float: Biegespannung in Pascal (Pa).
    """
    
    return moment * y / inertia
````

Der Docstring wird mit `"""` eingeleitet, es folgt eine kurze Erklärung der Funktion, anschließend eine Erklärung der Argumente und der Rückgabewerte, und zum Abschluss wieder `"""`. Wie genau der Docstring formatiert wird, hängt von der gewählten Konvention ab (wir haben hier den Google-Style verwendet). Viele Editoren greifen Docstrings in unterstützenden Funktionen auf (Tooltips oder Autovervollständigungen). Insbesondere kann nun aber jeder Entwickler und jede Entwicklerin über die Funktion `help()` eine Hilfe zu Ihrer Funktion aufrufen: 

````python
help(bending_stress)
````

Eine weitere Unterstützung bieten **Typ-Annotationen**, mit denen wir andere Entwickler bzw. Entwicklerinnen darüber informieren, welche Datentypen unsere Funktion erwartet bzw. zurückgibt. Aber Vorsicht: Diese Angaben sind rein informativ! Sie verhindern nicht, dass die Funktion mit anderen Datentypen aufgerufen wird! Unser Beispiel mit Typ-Annotationen: 

````python
def bending_stress(moment: float, y: float, inertia: float) -> float:
    """Berechnet die Biegespannung in einem Balken (Docstring wie oben)."""
    return moment * y / inertia
````

Die Typ-Annotationen geben an: Wir erwarten den Datentyp `float` für die Parameter `moment`, `y` und `inertia` und geben auch einen Float als Ergebnis zurück (`-> float`).

Schreiben Sie eine Funktion `ohms_law(R, U, I)`: Der Anwender übergibt für jeweils zwei der drei Parameter Zahlenwerte und für den dritten den String `"*"`. Daraufhin berechnet die Funktion den fehlenden Wert mithilfe des ohmschen Gesetzes und gibt ihn zurück. Dokumentieren Sie die Funktion mit einem Docstring, der die Benutzung erklärt.

````python
#|code-fold: true
def ohms_law(R, U, I):
    """
    Berechnet Widerstand, Spannung oder Strom mithilfe des ohmschen Gesetzes. 
    Der fehlende Wert wird mit dem Argument "*" markiert; für die anderen beiden müssen Zahlen übergeben werden.
    Args:
        R (float oder str): El. Widerstand (Ohm) oder "*".
        U (float oder str): El. Spannung (Volt) oder "*".
        I (float oder str): El. Strom (A) oder "*".
    Returns:
        float: fehlender Wert
    """
    
    if R == "*":
        return U / I
    elif U == "*":
        return R * I
    elif I == "*":
        return U / R
    else:
        print("Ungültige Eingabe! Siehe Hilfe: ")
        help(ohms_law)
````
