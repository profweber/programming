## Ein Sensor speichert stündlich Temperaturdaten – Anwendung der Liste
Ein Sensor misst stündlich die Temperatur (z.B. in einem Raum oder einer Maschine). Diese Daten speichern wir der Reihe nach in einer Liste, einer der einfachsten Datenstrukturen, die wir bereits im Kapitel über Datentypen kennengelernt haben. Die Liste ist ideal für lineares Speichern (Messwert für Messwert hintereinander). Sie ermöglicht den einfachen Zugriff auf Elemente anhand ihrer Position, stellt aber auch nützliche Funktionen zum Arbeiten mit den Daten bereit. Im Folgenden wiederholen wir einige wesentliche Eigenschaften von Listen und lernen weitere Möglichkeiten der Datenverarbeitung kennen.

### Speichern der Daten {.unnumbered}

````python
# eine Liste kann mit Werten initialisiert werden
temperatures = [22.5, 23.1, 21.8]

# es kann aber auch nach und nach ein Wert hinten angefügt werden:
temperatures.append(24.0)
temperatures.append(22.5)
temperatures.append(21.8)

print(temperatures)
````

In der Praxis würde an einer bestimmten Stelle im Code (einmal pro Stunde) mithilfe von `append` der aktuelle Messwert in der Liste gespeichert werden.

### Sortieren der Messungen {.unnumbered}
Sortieren hilft, die Werte geordnet darzustellen und z.B. das Finden des höchsten oder niedrigsten Messwerts zu erleichtern. Auch hierfür steht eine Methode bereit:

````python
# Sortieren der Liste
temperatures.sort()
print("Sortierte Temperaturen:", temperatures)
````

Wir können nun direkt ablesen, dass der Wertebereich zwischen 21.8 und 24.0 liegt. Allerdings haben wir unsere Originalliste durch das Sortieren verändert! Das war womöglich gar nicht unsere Absicht, denn wir wollen weiterhin die Information besitzen, wann welcher Wert erhoben wurde. Für diesen Zweck gibt es eine Lösung: Anstelle der Methode `.sort()` verwenden wir die bereits bekannte Built-in Function `sorted()` (siehe Kapitel *Funktionen*), die eine sortierte Liste zurückgibt, **ohne das Original zu verändern**:  

````python
# Wir stellen zunächst das Original wieder her
temperatures = [22.5, 23.1, 21.8, 24.0, 22.5, 21.8]

sorted_data = sorted(temperatures)
print("Sortierte Liste: ", sorted_data)
print("Original: ", temperatures)
````

Die implementierten Sortieralgorithmen sind übrigens sehr effizient (schneller als der früher vorgestellte Bubble Sort).

::: {.callout-important}

Beachten Sie unbedingt den Unterschied zwischen `.sort()` und `sorted()`:

`my_list.sort()  # Original verändert!`

`sorted_list = sorted(my_list)  # Original unverändert!`
:::

### Werte extrahieren – Indexzugriff und List Slicing {.unnumbered}

Wie bereits bekannt, kann man über Indizes auf die einzelnen Elemente zugreifen:

````python
print("1. Messwert: " , temperatures[0])
print("Letzter Messwert: " , temperatures[-1])
print("Min: " , sorted_data[0])
print("Max: " , sorted_data[-1])
print("5. Messwert: " , temperatures[4])
print("Vorletzter Messwert: " , temperatures[-2])
````

Oft interessiert man sich für ganze Bereiche von Messwerten und kann diese in Python sehr effizient mit sog. "List Slicing" *ausschneiden*: Hierfür verwendet man einen Start- und Endindex, getrennt durch ein `:`. Zu beachten ist, dass der Anfangsindex inkludiert wird, das Slicing aber einen Schritt vor dem Endindex aufhört (@fig-slicing):  

````python
# Slicing: Werte gesucht ...

# ... mit Indizes: 0,1,2
print(temperatures[0:3])  

# ... mit Indizes: 2,3,4
print(temperatures[2:5])  

# ... mit Indizes: 1,2,3,...,vorletztes
print(temperatures[1:-1])  
````

![Positive und negative Indizes der Liste `temperatures`. Beim Slicing gehört der Startindex dazu, der Endindex nicht.](images/slicing.png){#fig-slicing}

Man kann auch Start- und Endindex weglassen und meint damit "von Anfang" oder "bis Ende":

````python
# Slicing: Werte gesucht ...

# ... von Anfang bis Index 2
print(temperatures[:3]) 

# ... von Index 4 bis Schluss
print(temperatures[4:]) 

# Die ganze Liste als großer Slice:
print(temperatures[:]) 
````

Python erlaubt sogar noch präzisere Analysen, indem es Schrittweiten beim Slicing zulässt. Die Schrittweite ist standardmäßig `1`, aber kann als dritter Parameter hinter einem weiteren `:` verändert werden:

````python
# von Index 0 bis 4 in 2er-Schritten
print(temperatures[0:5:2]) 

# von Index 1 bis Schluss in 2er-Schritten
print(temperatures[1::2]) 

# jeder dritte Messwert
print(temperatures[::3]) 

# auch rückwärts funktioniert!
print(temperatures[4:1:-1]) 

# komplette Liste rückwärts
print(temperatures[::-1]) 
````

Man kann auch mit dem Operator `in` fragen, ob ein spezifischer Wert in der Liste enthalten ist:

````python
23.1 in temperatures
````

Dabei muss intern aber jedes Element einzeln abgefragt werden, mit Pech also die gesamte Liste durchgegangen werden (die Datentypen `set` und `dict` sind hier deutlich effizienter). 

Darüber hinaus sind die Built-in Functions `max()`, `min()`, `len()` oder `sum()` nützlich. 

````python
# Beispiel: Verwendung von Built-in Functions min, max, len
print("Min: ", min(temperatures))
print("Max: ", max(temperatures))
print("Anzahl der Elemente: ", len(temperatures))
````

## Analyse der verschiedenen Messwerte – Anwendung des Sets

Wir möchten wissen, welche verschiedenen Temperaturwerte vorkommen. Hierfür ist ein Set ideal, da es automatisch doppelte Einträge entfernt. Würden wir weiter mit einer Liste arbeiten, müssten wir selbst die Daten filtern, z.B. mit den uns bekannten Schleifen. Das wäre allerdings deutlich ineffizienter. Das Set entspricht einer mathematischen Menge, in der jedes Element nur einmal vorkommen darf. Für die Umwandlung führen wir ein explizites Casting durch:   

````python
unique_values = set(temperatures)
print(unique_values) # keine doppelten Werte mehr
````

Syntaktisch erkennen Sie das Set an den geschweiften Klammern `{}`. Beachten Sie: Ein Set ist ungeordnet. Die Reihenfolge der Ausgabe kann von der Eingabe abweichen, und einen Indexzugriff wie bei Listen gibt es nicht. Ein Set erlaubt alle bekannten Mengenoperationen, die Sie aus der Mathematik kennen (z.B. Schnittmenge oder Vereinigung). Nehmen wir an, wir wollen überprüfen, ob unser Sensor mit einer anderen Messreihe gemeinsame Werte hat. Hierfür können wir ganz einfach die Schnittmenge aus beiden Mengen untersuchen:  

````python
second_temperatures = [22.0, 23.1, 25.0, 22.4, 20.8]
second_set = set(second_temperatures)
intersection = unique_values.intersection(second_set)
print("Gemeinsame Werte: ", intersection)
````

Die Abfrage mit dem Operator `in` ist übrigens bei Sets viel effizienter als bei Listen. Das liegt an einem Trick: Zu jedem Element eines Sets berechnet Python einen sog. "Hash-Wert", der angibt, wo das Element im Speicher liegt. Fragt man, ob ein Element im Set liegt, muss Python nur den Hash-Wert für dieses Element berechnen und checken, ob der entsprechende Speicherplatz zum Set gehört. Daher ist der Aufwand - anders als bei Listen - unabhängig von der Größe des Sets! Deshalb dürfen Sets allerdings nur Elemente enthalten, für die sich ein Hash-Wert berechnen lässt (*hashable*). Das sind die unveränderlichen (immutable) Datentypen wie Zahlen, Strings oder Tupel (sofern das Tupel selbst keine Liste o. Ä. enthält); eine Liste als Element führt zu einem `TypeError`. Dasselbe gilt später für die Schlüssel von Dictionaries.

::: {.callout-tip}
### Wie initialisiert man ein Set? {.unnumbered}
Es gibt mehrere Möglichkeiten, ein neues Set anzulegen - leer oder mit Inhalt:

* Leere Initialisierung:	`my_set = set()` (**Achtung: leere geschweifte Klammern initialisieren ein Dictionary und kein Set!**)

* Mit Werten:	`my_set = {1, 2, 3}`

* Aus anderen Datenstrukturen: `my_set = set(x)`, wobei `x` einen String, eine Liste, ein Tupel oder ein Dictionary referenzieren kann. Im Falle des Strings erhält man beispielsweise die Menge der Zeichen (Duplikate werden entfernt!), im Falle des Dictionaries die Menge der Schlüssel.  
:::

::: {.callout-note}
### Wichtige Methoden des Sets {.unnumbered}
| Methode | Bedeutung |
|---------|-----------|
|`.add(x)`|Fügt das Element x hinzu.|
|`.remove(x)`|Entfernt x aus dem Set. (Achtung: Fehler, wenn x nicht vorhanden!)|
|`.clear()`| Entfernt alle Elemente.|
|`.union(*sets)`| Gibt die Vereinigungsmenge zurück (A ∪ B).|
|`.intersection(*sets)`|Gibt die Schnittmenge zurück (A ∩ B).|
|`.difference(*sets)`|Gibt die Differenzmenge zurück (A \\ B).|

Erinnerung: Darüber hinaus sind die Built-in Functions `max()`, `min()`, `len()`, `sum()`, `sorted()` (gibt eine sortierte Liste zurück) und der `in`-Operator nützlich!  
:::

## Hinzufügen von Zeitstempeln – Anwendung des Dictionaries

Für eine präzisere Darstellung ergänzen wir die Daten um Zeitstempel. Wir wären gerne in der Lage, folgende Frage zu beantworten: "Wie war die Temperatur am Tag xy zum Zeitpunkt z?". 

Hierzu speichern wir die Temperaturen als **Werte (values)** und die Zeitpunkte als **Schlüssel (keys)** in einem **Dictionary**. Bei einem Dictionary greift man über einen eindeutigen Schlüssel auf das entsprechende Element zu. Es bedarf keiner komplexen Suche nach dem passenden Element etc. Wir demonstrieren die Anwendung: 

````python
# Dictionary für Zeitstempel und Temperaturen
data = {
    "2025-03-23 08:00": 22.5,
    "2025-03-23 09:00": 23.1,
    "2025-03-23 10:00": 21.8,
    "2025-03-23 11:00": 24.0,
    "2025-03-23 12:00": 22.5,
}

# neue Werte können jederzeit einzeln hinzugefügt werden: 
data["2025-03-23 13:00"] = 21.8

print(data)
````

Achten Sie erneut auf die Syntax: Geschweifte Klammern (wie beim Set), aber darin befinden sich Auflistungen von `Key:Value`. Der Zugriff auf einzelne Elemente ähnelt syntaktisch dem Index-Zugriff bei einer Liste, aber anstelle eines Index verwendet man einen Schlüssel:   

````python
# Zugriff auf bestimmte Temperatur
print(data["2025-03-23 09:00"])
````

Eine Gefahr besteht darin, dass man auf Schlüssel zugreifen möchte, die es nicht gibt. Dann würde das Programm abstürzen! Daher gibt es noch eine sicherere Zugriffsmethode über die Methode `.get()`:

````python
print(data.get("2025-03-23 09:00"))
print(data.get("2025-03-23 22:00")) # Key existiert nicht
print(data.get("2025-03-23 22:00", "Datensatz fehlt")) # mit Alternativtext
````

::: {.callout-tip}
### Wie initialisiert man ein Dictionary? {.unnumbered}
Es gibt mehrere Möglichkeiten, ein neues Dictionary anzulegen - leer oder mit Inhalt:

* Leere Initialisierung:	`my_dict = {}` oder `my_dict = dict()`

* Mit Schlüssel-Wert-Paaren:	`my_dict = {"a": 1, "b": 2}`

* Aus einer Liste von Tupeln: `my_dict = dict([("a", 1), ("b", 2)])`
:::

::: {.callout-note}
### Wichtige Methoden des Dictionaries {.unnumbered}
| Methode | Bedeutung |
|---------|-----------|
|`.clear()`|Entfernt alle Elemente.|
|`.get(key, default)`|Gibt den Value zu `key` zurück; falls nicht vorhanden: `default`.|
|`.pop(key, default)`| Entfernt Eintrag zu `key` und gibt dessen Wert zurück; falls nicht vorhanden: `default`.|
|`.keys()` und `.values()`| Zum Iterieren über alle Keys bzw. Values.|
|`.items()`|Zum Iterieren über alle (key, value)-Paare.|

Erinnerung: Darüber hinaus sind die Built-in Functions `max()`, `min()`, `len()`, `sum()`, `sorted()` und der `in`-Operator nützlich, aber sie operieren standardmäßig auf den Keys des Dictionaries! Variante: `max(my_dict.values())`.  
:::

````python
# Beispiel zum Iterieren über alle Einträge: 
for time, temp in data.items():
    print(f"{time}: {temp}°C")
````

Das letzte Beispiel sieht etwas ungewohnt aus, da wir zwei Iterationsvariablen in einer `for`-Schleife gemeinsam verwenden: `time` und `temp`. In Wahrheit geschieht hier ein Entpacken von Tupeln (das Tupel wird im nächsten Abschnitt ausführlicher behandelt): Mit `.items()` iterieren wir über die Tupel `(key, value)` zu jedem Eintrag. Dieses Tupel haben wir aufgelöst in zwei Variablen, da wir wissen, dass die Keys den Zeitstempeln und die Values den Temperaturen entsprechen. Aber auch folgender Code funktioniert:  

````python
for t in data.items():
    print(f"key: {t[0]}, value: {t[1]}")
````

## Speichern einer zweiten Messgröße – Anwendung des Tupels (engl.: tuple)

Neben der Temperatur misst der Sensor nun auch die relative Luftfeuchtigkeit. Wir speichern beide Messwerte als **Tupel**. Ein Tupel ist ähnlich wie die Liste eine geordnete Kollektion von Elementen, aber mit einem wichtigen Unterschied: Das Tupel ist im Gegensatz zur Liste **unveränderlich**. Man kann weder Elemente hinzufügen noch entfernen oder überschreiben! Diese Einschränkungen machen das Tupel zu einem sicheren Datenformat, welches außerdem weniger Speicher verbraucht als die Liste. Syntaktisch erkennt man das Tupel meistens an den runden Klammern `()`:    

````python
measurement = (23.4, 40) # Tupel mit zwei Werten
print(measurement)
````

Die Klammern können in diesem Beispiel aber auch weggelassen werden:

````python
measurement = 23.4, 40 # ebenfalls ein Tupel mit zwei Werten (Merke: "das Komma macht das Tupel")
print(measurement)
````

Eine Funktion mit mehreren Rückgabewerten liefert ebenfalls ein Tupel:

````python
def my_function(x, y):
    return x + y, x - y, x * y

result = my_function(5, 3)  # Ergebnisse als Tupel speichern
print(result)
````

````python
# alternativ: Ergebnisse einzeln speichern
a, b, c = my_function(5, 3)  
print(a, b, c)
````

Man kann die Werte eines Tupels über den Indexzugriff auslesen (aber nicht überschreiben!): 

````python
print("Temperatur: ", measurement[0])
print("Luftfeuchtigkeit: ", measurement[1])

#measurement[0] = 25.0 ist nicht erlaubt! Ein Tupel ist unveränderlich!
````

Unser neues Dictionary sieht jetzt so aus: 

````python
# Erweiterung: Values = Tupel aus Temperatur und Luftfeuchtigkeit
data = {
    "2025-03-23 08:00": (22.5, 60),
    "2025-03-23 09:00": (23.1, 55),
    "2025-03-23 10:00": (21.8, 58),
    "2025-03-23 11:00": (24.0, 61),
    "2025-03-23 12:00": (22.5, 62)
}

# neuen Wert hinzufügen
data["2025-03-23 13:00"] = (21.8, 63)

# einen Zeitpunkt abfragen
print(data.get("2025-03-23 10:00"))

````

Tatsächlich könnte man die beiden Messwerte natürlich auch als Liste mit zwei Elementen zusammenfassen, aber das Tupel hat in diesem Fall klare Vorteile:

* Semantik: Ein Tupel repräsentiert eine feste Kombination von Werten, z.B. Temperatur und Luftfeuchtigkeit, die logisch zusammengehören.
* Unveränderbarkeit: Tupel verhindern unbeabsichtigte Änderungen. (Will man bewusst einen Wert verändern, kann man in eine Liste umwandeln.) 
* Effizienz: Tupel benötigen weniger Speicherplatz und werden (etwas) schneller verarbeitet als Listen.

::: {.callout-tip}
### Wie initialisiert man ein Tupel? {.unnumbered}
Es gibt mehrere Möglichkeiten, ein neues Tupel anzulegen - leer oder mit Inhalt:

* Leere Initialisierung:	`my_tuple = ()` oder `my_tuple = tuple()`

* Mit Werten:	`my_tuple = (1, 2, 3)`

* Ohne Klammern: `my_tuple = 1, 2, 3` wird automatisch als Tupel interpretiert. 

* Achtung Spezialfall - Tupel mit exakt einem Wert: `t = (42,)` oder `t = 42,` (Merke: *Das Komma macht das Tupel, nicht die Klammern!*) 

* Aus anderen Datenstrukturen: `my_tuple = tuple(x)`, wobei `x` einen String, eine Liste, ein Set oder ein Dictionary referenzieren kann. Im Falle des Strings erhält man beispielsweise das Tupel der Zeichen, im Falle des Dictionaries das Tupel der Schlüssel.  
:::

::: {.callout-note}
### Methoden des Tupels {.unnumbered}
Da das Tupel unveränderlich ist, stellt es nur Informationsmethoden zur Verfügung:

| Methode | Bedeutung |
|---------|-----------|
|`.count(x)`|Zählt das Vorkommen von x.|
|`.index(x)`|Gibt den Index des ersten Vorkommens von x zurück.|

Erinnerung: Darüber hinaus sind die Built-in Functions `max()`, `min()`, `len()`, `sum()`, `sorted()` und der `in`-Operator nützlich!  
:::

## Daten mehrerer Sensoren – verschachtelte Dictionaries

Stellen Sie sich vor, wir hätten drei anstelle von einem Sensor und wollten die Daten aller Sensoren strukturiert erfassen. Hierzu können wir zwei Dictionaries ineinander verschachteln: Der erste Key ist der Name des Sensors, zugehöriger Value ist ein Dictionary, das wie bisher die Messwerte dieses einen Sensors sammelt (also Zeitstempel als Key, Messwert als Value). Im Code sieht das so aus: 

````python
# Verschachtelte Dictionaries für drei Sensoren
data = {
    "Sensor1": {
        "2025-03-23 08:00": (22.5, 60),
        "2025-03-23 09:00": (23.1, 55)
    },
    "Sensor2": {
        "2025-03-23 08:00": (21.8, 57),
        "2025-03-23 09:00": (22.0, 67)
    },
    "Sensor3": {
        "2025-03-23 08:00": (24.0, 58),
        "2025-03-23 09:00": (23.8, 57)
    }
}

# So würde man Messwerte hinzufügen:
data["Sensor3"]["2025-03-23 10:00"] = (25.2, 62) # Neuer Wert für Sensor 3
````

## Mutable und immutable Datenstrukturen

Die Unterscheidung zwischen *mutable* und *immutable* Datentypen haben wir bereits in früheren Kapiteln kennengelernt. Es ist bei der Programmierung wichtig, jederzeit zu wissen, ob man mit einem mutable oder immutable Datentyp arbeitet. Bei den grundlegenden Datenstrukturen in Python lässt sich das zum Glück leicht merken: **Nur das Tupel ist unveränderlich (immutable)**, während Listen, Sets und Dictionaries mutable sind. Achtung: Enthält ein Tupel selbst veränderliche Elemente, z.B. eine Liste, kann sich deren Inhalt trotzdem ändern.

Alle mutable Datenstrukturen können potenziell Seiteneffekte verursachen. Im Folgenden illustrieren wir dies anhand von zwei Beispielen:

````python
# Beispiel: Seiteneffekte bei Sets
first_set = {"yellow", "red", "green"}
second_set = first_set
print("Identitäten gleich?", first_set is second_set)
second_set.add("blue")
print("First:", first_set, "Second:", second_set)
````

````python
# Beispiel: Seiteneffekte bei Dicts
first_dict = {"a": 1, "b" : 2}
second_dict = first_dict
second_dict["a"] = 42
print("First:",first_dict,"Second:",second_dict)
````

Und ein Tupel, das eine Liste enthält? Das Tupel selbst lässt sich nicht verändern, die Liste darin aber schon – zum Beispiel über eine zweite Referenz auf dieselbe Liste:

````python
lst = [1, 2]
t = (lst, 3)    # das Tupel enthält eine Referenz auf die Liste
lst.append(99)  # die Liste wird über die zweite Referenz lst verändert ...
print(t)        # ... und das Tupel "sieht" die Änderung
````

:::{.callout-important}
Listen, Sets und Dictionaries sind mutable, Seiteneffekte können auftreten!

Tupel sind immutable und verursachen selbst keine Seiteneffekte – solange sie keine veränderlichen Elemente wie Listen enthalten. 
:::

## Sicheres Kopieren von Listen

Seien Sie immer auf der Hut, wenn Sie eine Liste kopieren wollen. Da sie zu den mutable Datentypen gehört, erzeugen Sie schnell unbeabsichtigt nur eine zweite Referenz auf das Original mit den genannten Seiteneffekten als Gefahr. Eine einfache Möglichkeit, dies zu umgehen, ist eine Kopie mithilfe des Slicing-Operators: 

````python
original = ["apples", "oranges", "books"]

# Zuweisung erzeugt nur eine zweite Referenz:
second_reference = original  # keine Kopie! 

# mit Slicing eine Kopie erzeugen:
my_copy = original[:] 

# Wir zeigen, dass my_copy keine Seiteneffekte hat: 
my_copy.append("bananas")
print("Original: ", original) # unverändert
print("Kopie: ", my_copy)
````

Es gibt allerdings noch eine Diffizilität: Der Trick mit dem Slicing-Operator funktioniert nur so lange, wie die Elemente der Liste immutable sind. Sollten sich darunter aber mutable Datentypen befinden, kann es immer noch zu Seiteneffekten kommen. Wir haben zwar eine Kopie erstellt, allerdings nur eine sog. "flache Kopie". Um alle Mehrfachreferenzen aufzulösen und eine völlig unabhängige Kopie zu erhalten, braucht es eine sog. "tiefe Kopie". Hierfür gibt es bestimmte Algorithmen aus dem Modul `copy`:

````python
import copy
original = [[1,2],3,4] # Liste mit einem mutable Element

copy1 = original[:] # flache Kopie
copy2 = copy.deepcopy(original) # tiefe Kopie

# Demo der Seiteneffekte:
# Das mutable erste Element wird in den Kopien verändert
copy1[0].append(3) 
copy2[0].append(100)

print("Original: ", original)
print("flache Kopie: ", copy1)
print("tiefe Kopie: ", copy2)
````

Die Änderungen an der flachen Kopie hatten noch Auswirkungen auf das Original, weil beide die Referenz auf das erste Element teilten, welches mutable war. Nur bei der tiefen Kopie handelt es sich um eine völlig autarke Kopie, die nichts mehr mit dem Original gemein hat (@fig-copy-shallow-deep). 

![Flache Kopie: Die äußere Liste ist neu, die innere Liste wird aber geteilt. Tiefe Kopie: Auch die innere Liste wird kopiert.](images/copy_shallow_deep.png){#fig-copy-shallow-deep}

:::{.callout-tip}
### Vier Wege zur flachen Kopie {.unnumbered}
Die folgenden vier Varianten erzeugen alle eine **flache Kopie**:

1. Slicing: `my_copy = original[:]`
2. `list`-Funktion: `my_copy = list(original)`
3. Methode `.copy()`: `my_copy = original.copy()`
4. Funktion `copy()` aus Modul `copy`: `my_copy = copy.copy(original)`
:::

Die tiefe Kopie (`copy.deepcopy`) ist nicht als Built-in Function verfügbar, sondern muss aus dem Modul `copy` importiert werden. Das passt gut zur Python-Philosophie *„Explicit is better than implicit“*: Eine tiefe Kopie sollte nur dann eingesetzt werden, wenn man sie wirklich benötigt. Da die Operation rekursiv alle enthaltenen Objekte dupliziert, ist sie vergleichsweise aufwendig in Bezug auf Laufzeit und Speicherverbrauch. In vielen Fällen ist es daher sinnvoller, benötigte Datenstrukturen gezielt neu aufzubauen oder nur Teilstrukturen zu kopieren, anstatt pauschal eine tiefe Kopie der gesamten Struktur zu erzeugen.

## List Comprehensions

In Python gibt es eine elegante und kompakte Möglichkeit, Listen zu erzeugen: die sogenannte **List Comprehension**.

Statt mit einer Schleife umständlich Elemente einer Liste hinzuzufügen, können wir in einer einzigen Zeile eine neue Liste erstellen.
Das macht den Code kürzer, lesbarer und oft schneller.

### Syntax {.unnumbered}

Die allgemeine Syntax sieht so aus: 

`[ Ausdruck for Element in Sequenz ]`

- **Sequenz**: Etwas, über das iteriert werden kann (z.B. eine Liste, ein String, ein Bereich `range()`), wie wir es schon von der for-Schleife kennen.

- **Element**: Eine Variable, die nacheinander die Werte aus der Sequenz annimmt.

- **Ausdruck**: Ein Ausdruck, der für jedes Element ausgewertet wird. Dieser wird dann der neuen Liste hinzugefügt.

Der hintere Teil der List Comprehension verhält sich ganz analog zur for-Schleife, neu ist der Ausdruck. Ein Beispiel macht alles verständlicher:

````python
# Beispiel: Liste aller Quadratzahlen der Zahlen von 0 bis 9
squares = [x**2 for x in range(10)]
print(squares)
````

Analysieren wir das Beispiel:

- **Sequenz**: `range(10)`. 

- **Element**: `x`, eine eigene Bezeichnung für die Elemente der Sequenz.

- **Ausdruck**: `x**2`, jedes `x` aus der genannten Sequenz soll quadriert werden.

Die Syntax kann noch um eine `if`-Bedingung am Ende erweitert werden: 

`[ Ausdruck for Element in Sequenz if Bedingung ]`

- **Bedingung** am Ende der List Comprehension: Nur Elemente, die die Bedingung erfüllen, werden berücksichtigt.

````python
# Ein Beispiel: Liste aller Quadratzahlen der geraden Zahlen von 0 bis 9
squares2 = [x**2 for x in range(10) if x % 2 == 0]
print(squares2)
````

Darüber hinaus können auch `if`- und `else`-Bedingungen vor dem `for` verwendet werden, um alternative Ausdrücke zu verwenden (vor dem `for` ist das `else` Pflicht; ein reiner Filter ohne `else` gehört ans Ende): 

````python
# Mit if-else vor dem for
result = [x if x % 2 == 0 else -x for x in range(10)]
print(result)
````

Je komplexer eine List Comprehension wird, desto schwieriger wird sie allerdings zu lesen und zu verstehen – und das kann schnell die Vorteile zunichtemachen, die man eigentlich durch die kürzere Schreibweise erreichen wollte.

### Beispiele aus der Ingenieurpraxis {.unnumbered}

List Comprehensions sind sehr praktisch, um Messwerte von einer Einheit in eine andere umzurechnen:

````python
values_in_cm = [100, 250, 430, 125]
values_in_m = [x / 100 for x in values_in_cm]
print(values_in_m)
````

````python
temperatures_celsius = [0, 20, 37, 100]
temperatures_fahrenheit = [t * 9/5 + 32 for t in temperatures_celsius]
print(temperatures_fahrenheit)
````

List Comprehensions werden auch gerne zum Filtern von Messwerten verwendet, wie das folgende Beispiel zeigt:

````python
voltages = [3.3, 5.1, 2.5, 4.8, 2.1, 7.2, 0.7]
high_voltages = [u for u in voltages if u > 5]  # 5 V ist hier ein oberer Grenzwert 
low_voltages = [u for u in voltages if u <= 5]
print("Oberhalb des Grenzwertes:", high_voltages)
print("Gültige Messwerte:", low_voltages) 
````

````python
# Alternativ könnte man die zu hohen Werte "abschneiden":
voltages = [3.3, 5.1, 2.5, 4.8, 2.1, 7.2, 0.7]
limited_voltages = [u if u <= 5 else 5 for u in voltages]  # 5 V ist hier ein oberer Grenzwert 
print("Bereinigte Messwerte:", limited_voltages) 
````

### Fallstrick: Elemente entfernen, während man über die Liste iteriert {.unnumbered}
```python
voltages = [4.8, 5.3, 5.6, 4.9]
for u in voltages:
    if u > 5.0:
        voltages.remove(u)
print(voltages)   # [4.8, 5.6, 4.9] – 5.6 wurde übersehen!
```

Die Schleife merkt sich intern, an welcher Position sie gerade ist. Nach dem Entfernen von 5.3 rutscht 5.6 auf dessen Platz nach. Die Schleife macht aber an der nächsten Position weiter und überspringt 5.6 dadurch.

**Lösung:** Die Schleife wird **komplett ersetzt**. Statt Elemente aus der alten Liste zu entfernen, baut man eine **neue** Liste, die nur die gewünschten Elemente enthält:

```python
voltages = [4.8, 5.3, 5.6, 4.9]
voltages = [u for u in voltages if u <= 5.0]   # ersetzt die ganze Schleife
print(voltages)   # [4.8, 4.9]
```

Ausgeschrieben als Schleife sieht dasselbe so aus:

```python
voltages = [4.8, 5.3, 5.6, 4.9]
valid = []
for u in voltages:          # über die alte Liste iterieren ...
    if u <= 5.0:
        valid.append(u)     # ... aber nur die neue Liste verändern
voltages = valid
```

### Fallstrick: Verschachtelte Listen mit `*` erzeugen {.unnumbered}
```python
matrix = [[0] * 3] * 3   # sieht aus wie eine 3x3-Matrix ...
matrix[0][0] = 1
print(matrix)
# Ausgabe: [[1, 0, 0], [1, 0, 0], [1, 0, 0]] – alle Zeilen geändert!
```

**Was passiert hier?** Python wertet den inneren Ausdruck `[0] * 3` **genau einmal** aus. Dabei entsteht eine einzige Liste `[0, 0, 0]`. Das äußere `* 3` kopiert diese Liste nicht, sondern vervielfältigt nur die **Referenz** auf sie. Das ist wie bei `b = a`, das ja auch nie kopiert. Die Zeile bedeutet also dasselbe wie:

```python
row = [0] * 3
matrix = [row, row, row]       # dreimal dieselbe Liste
print(matrix[0] is matrix[1])  # True
```

**Richtig:** `matrix = [[0] * 3 for i in range(3)]`

Der Unterschied (@fig-matrix-references): Die List Comprehension führt den Ausdruck `[0] * 3` in **jedem Durchlauf neu** aus. So entstehen drei verschiedene Listen, genau wie bei dieser Schleife:

```python
matrix = []
for i in range(3):
    matrix.append([0] * 3)     # in jedem Durchlauf eine neue Liste

print(matrix[0] is matrix[1])  # False
matrix[0][0] = 1
print(matrix)                  # [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
```

![Links: `* 3` vervielfältigt nur die Referenz auf eine einzige Zeile. Rechts: Die List Comprehension erzeugt drei verschiedene Listen.](images/matrix_references.png){#fig-matrix-references}

Auch `[0] * 3` vervielfältigt übrigens nur Referenzen, nämlich auf die Zahl 0. Das ist aber harmlos: `int` ist immutable, und `matrix[0][0] = 1` lässt nur diesen einen Platz auf ein neues Objekt zeigen.

### Performance {.unnumbered}

List Comprehensions sind in der Regel schneller als eine klassische Schleife mit `append()`, weil Python dabei intern einige Schritte einsparen kann, vor allem den wiederholten Aufruf der Methode `append()`. Eine kleine Demo:

````python
import time  # zum Messen von Zeiten

# Klassische For-Loop mit append
start = time.time()

result = []
for i in range(10000000):
    if i % 2 == 0:
        result.append(i * 2)
end = time.time()

print(f"For-Loop Dauer: {round(end - start,2)} Sekunden")

# List Comprehension mit demselben Ergebnis

start = time.time()

result = [i * 2 for i in range(10000000) if i % 2 == 0]

end = time.time()

print(f"List Comprehension Dauer: {round(end - start,2)} Sekunden")
````

Die Zeiten variieren natürlich und werden auf Ihrem System andere Werte annehmen. Wie groß der Unterschied ausfällt, hängt von Python-Version und Rechner ab; oft ist er geringer als hier. Probieren Sie es aus!

Ob ein Code, wie in diesem Beispiel, eine Sekunde oder zwei Sekunden benötigt, spielt natürlich keine große Rolle. Der Performance-Unterschied wird nur relevant, wenn Sie enorm große Mengen an Daten verarbeiten müssen. 

### Fazit für die Praxis {.unnumbered}

List Comprehensions sind oft eine gute Wahl, wenn eine neue Liste erzeugt werden soll. Sie führen in der Regel zu kürzerem und verständlicherem Code. Allerdings sollte man sie nur verwenden, wenn Bedingungen und Ausdrücke einfach genug sind, um die Lesbarkeit nicht zu gefährden. Bei komplexerer Logik ist eine klassische Schleife oft die bessere Wahl, um die Wartbarkeit zu sichern. Sollte Performance eine entscheidende Rolle spielen, kann es sich lohnen, größere Schleifen in eine List Comprehension umzuschreiben – vorausgesetzt, die Lesbarkeit bleibt erhalten.

## Eleganter iterieren: `enumerate()` und `zip()` {#sec-enumerate-zip}

In Schleifen braucht man oft mehr als nur das aktuelle Element: seine **Position** in der Liste oder das **passende Element einer zweiten Liste**, etwa den Zeitpunkt zu einem Messwert. Mit `range(len(...))` und Indexzugriffen geht das zwar, Python bietet dafür aber zwei eingebaute Funktionen, die den Code kürzer und lesbarer machen.

`enumerate()` liefert bei jedem Schleifendurchlauf ein Tupel aus **Index und Element**, das wir – wie bei `.items()` – direkt in zwei Variablen entpacken:

````python
temperatures = [21.4, 23.9, 27.5, 31.2, 36.8]

for i, temp in enumerate(temperatures):
    print(f"Messung {i}: {temp} °C")
````

`zip()` fügt mehrere Sequenzen zusammen wie ein Reißverschluss – daher der Name: Bei jedem Durchlauf erhalten wir ein Tupel mit den Elementen, die in den Sequenzen an derselben Position stehen:

````python
times_min = [0, 5, 10, 15, 20]

for t, temp in zip(times_min, temperatures):
    print(f"t = {t} min: {temp} °C")
````

### `zip()` bei unterschiedlich langen Sequenzen {.unnumbered}
