## Grundidee und Prinzipien

Die objektorientierte Programmierung (OOP) ist ein Programmierparadigma, das Software in Form von **Objekten** strukturiert. Diese Objekte kombinieren **Daten (Attribute)** und **Verhalten (Methoden)** in einer einzigen Einheit. Das Ziel ist es, die reale Welt besser im Code abzubilden, indem Software-Entitäten geschaffen werden, die Eigenschaften und Fähigkeiten besitzen. Dadurch können sehr komplexe Systeme übersichtlich modelliert werden. OOP fördert dadurch eine modularere und wartungsfreundlichere Softwareentwicklung, die näher an der Wirklichkeit ist.

Vier zentrale Prinzipien prägen die objektorientierte Programmierung:

* **Kapselung (Encapsulation)**

    Daten und Funktionen, die zusammengehören, werden in **Klassen** zusammengefasst. Der Zugriff auf sensible Daten wird kontrolliert – häufig durch private Attribute und öffentliche Schnittstellen (Methoden), die genau festlegen, wie die Attribute verändert oder gelesen werden dürfen.
Eine Klasse ist die abstrakte Beschreibung einer Gesamtheit: Sie definiert, was Objekte dieser Art auszeichnet. Beispielsweise beschreibt die Klasse *Person* abstrakt Eigenschaften wie Name, Geburtsdatum oder Wohnort. Eine konkrete Instanz, wie etwa Max Mustermann, ist dann ein Objekt dieser Klasse.

* **Vererbung (Inheritance)**

    Klassen können Eigenschaften und Verhalten von anderen Klassen übernehmen, um Redundanzen zu vermeiden und Code-Wiederverwendung zu ermöglichen. Man kann sich Klassen vorstellen, die verschiedene Tierarten beschreiben, etwa *Dog*, *Cat* oder *Mouse*. Obwohl sich diese Klassen unterscheiden, teilen sie gemeinsame Merkmale, die in einer Basisklasse *Animal* zusammengefasst werden können. Die abgeleiteten Klassen spezifizieren dann die Besonderheiten jeder Art. Man sagt: "*Dog* erbt von *Animal*" oder "*Dog* ist abgeleitet von *Animal*".

* **Polymorphie oder Polymorphismus (Polymorphism)**

    Objekte verschiedener Klassen können über die gleiche Schnittstelle angesprochen werden. Dadurch können Funktionen allgemeiner und flexibler gestaltet werden.
Im obigen Beispiel kann eine Funktion, die ein Objekt der Klasse *Animal* erwartet, gleichermaßen mit einem *Dog*, einer *Cat* oder einer *Mouse* arbeiten. Für die Funktion ist nur relevant, dass es sich um ein Animal-Objekt handelt, nicht um dessen genaue Art. Polymorphismus unterstützt dadurch die Erweiterbarkeit und Anpassbarkeit von Programmen. Beispiel: In einer Tierarztpraxis werden alle Patienten in einer Datenbank verwaltet. Dabei ist es egal, ob es sich um Hunde, Katzen oder Mäuse handelt – alle Tiere werden als Instanzen der allgemeinen Klasse *Animal* gespeichert. Die spezifischen Eigenschaften (z.B. Hunderasse oder Lieblingsfutter der Katze) werden in abgeleiteten Klassen ergänzt.


* **Abstraktion (Abstraction)**
  
    Nur die wesentlichen Merkmale eines Objekts werden dargestellt; unwichtige Details werden ausgeblendet. Dies erleichtert die Handhabung komplexer Systeme.
Dieses Prinzip entspricht dem allgemeinen Konzept der Modellbildung: Ein Modell sollte so präzise wie nötig, aber so einfach wie möglich sein. Eine vollständige Abbildung der Realität wäre genauso komplex wie die Realität selbst und damit unpraktisch. Daher ist Abstraktion immer ein bewusster Kompromiss – zugunsten besserer Verständlichkeit und Lösbarkeit von Problemen. In der objektorientierten Programmierung kann man sogar sog. abstrakte Klassen programmieren, die zwar selbst nicht instanziiert werden können, aber die Struktur aller abgeleiteten Klassen festlegen.

### Paradigmen der Programmierung {.unnumbered}
In der Softwareentwicklung bezeichnet der Begriff Paradigma einen grundlegenden Ansatz oder eine Methode, wie Probleme durch Programmieren gelöst werden. Jedes Paradigma hat seine eigenen Prinzipien und Konzepte, die eine bestimmte Herangehensweise an die Problemstellung fördern. Die Wahl des Paradigmas beeinflusst die Struktur des Codes, die Art der Probleme, die gelöst werden können, und wie effizient der Entwicklungsprozess verläuft. Jedes Programmierparadigma hat seine Stärken und eignet sich für bestimmte Arten von Aufgaben. In der Praxis wird häufig eine Kombination verschiedener Paradigmen verwendet.

Wichtige Programmierparadigmen:

1. **Imperative Programmierung**:

    In der imperativen Programmierung wird die Ausführung einer Software durch eine Reihe von Anweisungen beschrieben, die den Zustand des Programms verändern. Der Fokus liegt darauf, wie ein Problem gelöst wird, also auf den einzelnen Schritten und deren Reihenfolge. Prozedurale Programmierung erweitert die imperative Programmierung um die Möglichkeit, Code in wiederverwendbare Teilaufgaben zu zerlegen. Praktisch alle aktuellen imperativen Programmiersprachen unterstützen den prozeduralen Ansatz. 
   
    Hauptprinzipien: Sequentielle Anweisungen, Zustandsänderung, Schleifen und Bedingungen.

    Beispiele: Sprachen wie C, Python und Java unterstützen dieses Paradigma.
    
2. **Objektorientierte Programmierung (OOP)**:

    In der objektorientierten Programmierung werden Daten und Funktionen, die auf diese Daten operieren, in Objekten organisiert. OOP fördert die Wiederverwendbarkeit und Strukturierung von Code durch die Konzepte von Klassen, Objekten, Vererbung und Polymorphismus.

    Hauptprinzipien: Kapselung, Vererbung, Polymorphismus, Abstraktion.
    
    Beispiele: Java, Python, C++.

3. **Funktionale Programmierung**:

    Funktionale Programmierung basiert auf der Verwendung von Funktionen als grundlegende Bausteine des Programms. Sie betont die Unveränderlichkeit von Daten und vermeidet Zustandsänderungen und Seiteneffekte. Funktionen werden als *First-Class-Objekte* behandelt, was bedeutet, dass sie als Argumente übergeben und als Rückgabewerte genutzt werden können.

    Hauptprinzipien: Unveränderliche Daten, Funktionen als Objekte, keine Seiteneffekte.
    
    Beispiele: Haskell, Scala.

4. **Logische Programmierung**:

    Bei der logischen Programmierung wird ein Problem in Form von Fakten und Regeln beschrieben, die zusammen die Lösung des Problems definieren. Während der Programmierung gibt man an, was wahr ist, und der Computer zieht Schlussfolgerungen, um zu einer Lösung zu gelangen.

    Hauptprinzipien: Fakten, Regeln, Schlussfolgerungen.
   
    Beispiel: Prolog.
:::

## Bestandteile einer Klasse: Methoden und Attribute

Ohne genauer darauf hinzuweisen, haben wir bereits sehr oft mit Objekten gearbeitet: In Python 3 sind nämlich alle Datentypen in Wahrheit Klassen. Besonders auffällig ist dies bei der Verwendung von Datenstrukturen, wie Listen: 

```python
my_list = [3, 1, 2]  # erzeugt ein Objekt der Klasse "list"
print(type(my_list))
```

Die Klasse `list` bietet eine Vielzahl an **Methoden** an, das sind alle Funktionen, die innerhalb der Klasse definiert sind, z.B.: 

```python
my_list.sort()  # sort-Methode der Klasse list
my_list.reverse()  # reverse-Methode
my_list.append(10)  # append-Methode
```

Wir betrachten nun als weiteres Beispiel die Klasse `datetime` aus dem gleichnamigen Modul:

```python
from datetime import datetime  # importiert die Klasse

# wir erzeugen eine Instanz von datetime 
# mit dem heutigen Datum:
today = datetime.now()
```

Über die Referenz `today` können wir nun die erzeugte Instanz untersuchen und auf ihre Eigenschaften - die sogenannten **Attribute** -  zugreifen:

```python
today.year  # Zugriff auf das Attribut year
```

```python
today.month  # Zugriff auf das Attribut month
```

Die Klasse bietet aber auch Methoden an, z.B. `toordinal()`, eine Funktion, die das Datum als Integer ausgibt, der die Tage fortlaufend nummeriert, wobei der 1. Januar des Jahres 1 n. Chr. die Nummer 1 hat:

```python
today.toordinal()
```

Ein zweites Beispiel ist die Methode `strftime`, mit der ein Datum als formatierter String ausgegeben wird:

```python
today.strftime("%d.%m.%Y")  # das Format ist beliebig anpassbar
```

Als wir unsere Instanz oben erzeugt haben, verwendeten wir die Methode `now()`. Diese Methode konnten wir ohne konkretes Objekt verwenden: Sie wurde direkt über den Klassennamen `datetime` aufgerufen und liefert ein neues `datetime`-Objekt. Methoden, die zur Klasse selbst gehören und keine Instanz benötigen, heißen in vielen Programmiersprachen **statische Methoden**. Python unterscheidet hier genauer zwischen **Klassenmethoden** (sie erhalten die Klasse als ersten Parameter; `now()` ist eine solche) und **statischen Methoden** (sie erhalten weder Instanz noch Klasse). Entsprechend gibt es **Klassenattribute**, die zur Klasse gehören und von allen Instanzen geteilt werden.  

Fassen wir die wichtigsten Begriffe zusammen:

::: {.callout-note appearance="minimal"}

**Attribut**

Ein Attribut ist eine Variable innerhalb einer Klasse, die den Zustand oder eine Eigenschaft eines Objekts beschreibt.

Beispiel: `today.year`, wobei `today` ein Objekt der Klasse `datetime` referenziert.

**Methode**

Eine Methode ist eine Funktion innerhalb einer Klasse, die das Verhalten eines Objekts definiert und auf dessen Attribute zugreifen oder sie verändern kann.

Beispiel: `today.toordinal()`, wobei `today` ein Objekt der Klasse `datetime` referenziert.

**Statisch**

Statisch bezeichnet Elemente, die zur Klasse selbst gehören und nicht an einzelne Objekte gebunden sind. Sie werden unabhängig von Instanzen verwendet. In Python spricht man genauer von Klassenattributen sowie von Klassenmethoden und statischen Methoden.

Beispiel: `datetime.now()` (in Python eine Klassenmethode)

**Members**

Der Begriff **Members** fasst alle Bestandteile einer Klasse zusammen; alle Attribute und alle Methoden (auch statische) sind Members der Klasse.
:::

## Erste eigene Klassen

Wir programmieren erste kleine Klassen, um uns an das Konzept und die Syntax zu gewöhnen.

```python
class Animal:
    
    # mit der Methode __init__ - dem "Konstruktor" - fängt alles an:
    def __init__(self, name, age):
        self.name = name  # Attribut name wird angelegt
        self.age = age  # Attribut age wird angelegt

    # eine weitere Methode:
    def speak(self):
        print(f"{self.name} macht ein Geräusch.")

    # eine weitere Methode:
    def birthday(self):
        print(f"Herzlichen Glückwunsch zum Geburtstag, {self.name}!")
        self.age += 1
```

Wir erkennen die allgemeine Syntax zum Definieren einer Klasse:

```python
#| output: false
class Klassenname:
    # eingerückter Code
    # Definition aller Methoden
    ...
```

Der Klassenname soll nach den offiziellen Empfehlungen mit einem Großbuchstaben beginnen und bei mehreren Wörtern in CapWords-Schreibweise (auch PascalCase genannt, siehe PEP 8) geschrieben werden (z.B. `GameController` oder `WindowManager`). 
Im obigen Beispiel haben wir die Klasse `Animal` definiert und drei Methoden angelegt: `__init__(...)`, `speak(...)` und `birthday(...)`. Auffällig ist, dass alle Methoden unserer Klasse als erstes Argument den Parameter `self` verwenden. Dieser Parameter verweist auf das aktuelle Objekt der Klasse. Das bedeutet, dass `self` die Instanz der Klasse repräsentiert, die die Methode gerade aufruft. Der Parameter wird beim Aufruf automatisch gesetzt: `my_dog.speak()` (siehe unten) ist gleichbedeutend mit `Animal.speak(my_dog)`, `self` ist dann also das Objekt `my_dog`.

Ein weiteres auffälliges Merkmal ist die Methode `__init__`, die durch die doppelten Unterstriche vor und nach ihrem Namen hervorsticht. Methoden dieser Art werden in Python als **Magic Methods** bezeichnet, da sie eine spezielle Bedeutung für den Python-Interpreter haben. In diesem Fall ist `__init__` der **Konstruktor** einer Klasse. Der Konstruktor wird automatisch aufgerufen, wenn ein neues Objekt dieser Klasse erstellt wird. Neben dem obligatorischen `self` erhält der Konstruktor die Argumente `name` und `age`, die dann über die Zuweisungen `self.name = name` und `self.age = age` als Attribute in der Instanz gespeichert werden. Die Übergabeparameter des Konstruktors müssen dabei nicht zwangsläufig genauso heißen wie die Attribute, allerdings ist das eine weit verbreitete Praxis, um Verwechslungen bei der Zuweisung zu vermeiden.

Die Methoden `speak` und `birthday` sind relativ selbsterklärend: `speak` gibt eine allgemeine Meldung des Tieres auf dem Bildschirm aus (mit `print`; einen Rückgabewert hat die Methode nicht), während `birthday` zum Geburtstag gratuliert und das Alter um eins erhöht. Testen wir nun unsere Klasse:

```python
my_dog = Animal("Bello", 5)  # ruft den Konstruktor auf
my_dog.speak()
```

```python
my_dog.age  # greift auf das Attribut age zu
```

```python
my_dog.birthday()
```

```python
my_dog.age
```

```python
# Natürlich können wir beliebig viele
# Animals erzeugen, hier ein zweites:
my_cat = Animal("Liz", 1)
my_cat.speak()
```

Die `speak`-Methode ist noch sehr unzufriedenstellend. Ein Hund sollte doch bellen, eine Katze miauen. Allerdings ist `Animal` eine relativ abstrakte Basisklasse für alle Tiere und deshalb an dieser Stelle zwangsläufig unspezifisch. Wir können nun das Prinzip der Vererbung verwenden, um spezifische Tierklassen anzulegen: 

```python
# Vererbung: Dog erbt von Animal
class Dog(Animal):
    def speak(self):
        print(f"{self.name} bellt.")

# Vererbung: Cat erbt von Animal
class Cat(Animal):
    def speak(self):
        print(f"{self.name} miaut.")
```

Die Basisklasse (hier: `Animal`) wird bei der Klassendefinition in Klammern angegeben. Das Anlegen der Klassen `Dog` und `Cat` ist sehr einfach, denn wir spezifizieren nur die Unterschiede zur Basisklasse, also die individuellen Realisierungen der Methode `speak`. Alles andere wird von der Basisklasse `Animal` übernommen (die abgeleiteten Klassen werden deshalb auch "Subklassen" oder "Kindklassen" genannt). Probieren wir unsere neuen Klassen aus: 

```python
my_dog = Dog("Bello", 6)
my_cat = Cat("Liz", 1)
my_turtle = Animal("Einstein", 100)

my_dog.speak()
my_cat.speak()
my_turtle.speak()
```

Man sagt auch: Die Methode `speak` wurde in den Klassen `Dog` und `Cat` *überschrieben*. Mit Vererbung könnten wir noch deutlich spezifischere Klassen anlegen, z.B. konkrete Hunderassen (hierzu würden wir von `Dog` ableiten, z.B. `class Husky(Dog):`). Es entstehen beliebig komplexe Klassen-Hierarchien. An unserem Beispiel erkennen wir auch, dass die Basisklasse davon unbeschadet ist und weiterhin verwendet werden kann (`my_turtle` referenziert eine allgemeine `Animal`-Instanz).   

Alle Subklassen von `Animal` werden weiterhin als `Animal` betrachtet und können auch als solche verwendet werden. Das bedeutet, dass eine Funktion, die ein `Animal` erwartet, auch mit einem `Dog` oder einer `Cat` funktioniert. Dies ist ein Beispiel für Polymorphismus. Wichtig: Obwohl alle Instanzen Tiere sind, sind sie nicht alle gleich - die gleiche Methode (hier `speak()`) wird je nach konkretem Typ unterschiedlich ausgeführt.

Umgekehrt muss man vorsichtiger sein: Eine Subklasse kann zusätzliche Methoden oder Attribute haben, die in der Basisklasse nicht existieren. Wenn eine Funktion speziell auf diese neuen Methoden zugreifen möchte, muss sie den spezifischen Typ des Objekts kennen, anstatt nur die Basisklasse zu verwenden. Andernfalls könnte der Zugriff auf diese zusätzlichen Eigenschaften oder Methoden zu einem Fehler führen, falls das Objekt tatsächlich eine Instanz der Basisklasse und nicht der Subklasse ist. Beispielhaft erweitern wir unsere Klasse `Dog` um eine Methode `fetch()`:

```python
class Dog(Animal):
    def speak(self):
        print(f"{self.name} bellt.")

    def fetch(self):
        print(f"{self.name} fängt den Ball.")
```

```python
my_dog = Dog("Bello", 6)
my_dog.fetch()

my_turtle = Animal("Einstein", 100)
# my_turtle.fetch() FEHLER! Die Methode existiert nicht in der Basisklasse!

my_cat = Cat("Liz", 1)
# my_cat.fetch() FEHLER! In der Klasse Cat existiert sie auch nicht!
```

## UML-Klassendiagramme

Wir wollen uns nun mit dem systematischen Design von Klassen befassen. Häufig wird dazu die **Unified Modeling Language (UML)** verwendet. Es handelt sich dabei um eine standardisierte grafische Sprache zur Modellierung und Visualisierung von Software-Systemen. UML ermöglicht es Entwickler:innen, die Struktur, das Verhalten und die Interaktionen eines Systems darzustellen. Ein besonders wichtiger Bestandteil von UML ist das Klassendiagramm. Es stellt die statische Struktur eines Systems dar, indem es die Klassen eines Systems sowie deren Beziehungen zueinander visualisiert. UML ist sehr mächtig und kann hier nur in kleinen Auszügen dargestellt werden. 

### Grundstruktur eines Klassendiagramms {.unnumbered}

Eine Klasse wird durch ein Rechteck dargestellt, welches in drei übereinanderliegende Felder unterteilt wird (siehe @fig-uml-class): 

- Der Name der Klasse wird in das oberste Feld geschrieben.
- Im zweiten Feld werden alle Attribute aufgeführt.
- Im dritten Feld folgen die Methoden.

![Aufbau eines UML-Klassendiagramms.](images/oop-uml-class.png){#fig-uml-class width=30%}

Konkret sieht es für unsere `Animal`-Klasse aus dem vorigen Abschnitt so aus wie in @fig-uml-animal.

Vererbung wird im UML-Diagramm durch einen Pfeil mit nicht ausgefüllter Dreiecksspitze dargestellt, der **von der abgeleiteten Klasse zur Basisklasse** zeigt. In der abgeleiteten Klasse werden Attribute und Methoden, die unverändert von der Basisklasse übernommen werden, nicht aufgeführt. In @fig-uml-inheritance sehen wir dies für die Klassen `Dog` und `Animal`.

#### Zugriffsmodifikatoren {.unnumbered}

Die `+`-Zeichen in den bisherigen Diagrammen sind nicht beliebig, sondern stellen sog. **Zugriffsmodifikatoren** dar. Sie geben an, wer über die gekennzeichneten Attribute und Methoden verfügen darf. Man unterscheidet drei Stufen: 

* **Public (`+`)**: Öffentliche Elemente, die von überall und jedem aufgerufen werden können.

* **Private (`-`)**: Private Elemente, die nur innerhalb der Klasse verwendet werden können. Außenstehende sollen diese Elemente nicht verwenden. Auch abgeleitete Klassen sollen keinen Zugriff auf sie haben.  

* **Protected (`#`)**: Geschützte Elemente, die in der Klasse und ihren Unterklassen verwendet werden können. Für Außenstehende sollen sie aber nicht direkt verfügbar sein.

In Python gibt es keine Möglichkeit, die Zugriffsrechte eins zu eins umzusetzen. Das ist ein großer Unterschied zu anderen Programmiersprachen, wie z.B. C++. Dennoch gibt es Mechanismen und Konventionen, wie in Python mit den unterschiedlichen Zugriffsrechten umgegangen wird. Darauf gehen wir in einem späteren Abschnitt genauer ein.   

### Beziehungen zwischen Klassen {.unnumbered}

In UML wird eine ungerichtete Verbindungslinie verwendet, um auszudrücken, dass zwei Klassen miteinander in Beziehung stehen. Diese Beziehung wird als *Assoziation* bezeichnet. Ein Beispiel hierfür ist die Beziehung zwischen den Klassen `Person` und `Animal` in @fig-uml-association, wobei eine Person Haustiere besitzen kann. In UML wird in diesem Fall häufig die **Multiplizität** angegeben, um zu definieren, wie viele Instanzen der einen Klasse mit wie vielen Instanzen der anderen Klasse in Beziehung stehen können.

In unserem Beispiel kann eine Person 0 oder beliebig viele Tiere besitzen, während jedes Tier exakt eine Person als Besitzer bzw. Besitzerin hat (eine vereinfachte Annahme). Dies verdeutlicht, dass zu jeder Assoziation immer zwei Multiplizitäten gehören, die direkt an den Enden der Verbindungslinie notiert werden. Die gängigsten Bezeichner für Multiplizitäten sind:

1: genau eine Instanz.

0..1: keine oder genau eine Instanz.

0..*: keine oder beliebig viele Instanzen.

1..*: eine oder beliebig viele Instanzen.

m..n: mindestens m Instanzen, maximal n Instanzen (z.B. 3..10).

![UML-Klassendiagramm mit Assoziation: Beziehung zwischen `Person` und `Animal`.](images/oop-uml-association.png){#fig-uml-association width=60%}

### Ergänzungen {.unnumbered}

* Im UML-Klassendiagramm kann außerdem der Datentyp von Attributen mitangegeben werden. Dazu verwendet man einen Doppelpunkt hinter dem Attribut gefolgt von der Typangabe, z.B. `name: str`.
* Bei den Methoden werden in den Klammern alle Parameter angegeben, allerdings ohne den Parameter `self`. Dieser muss in Python zwar in jeder Methodendefinition ausdrücklich als erster Parameter stehen (Ausnahmen: Klassenmethoden erhalten stattdessen die Klasse, statische Methoden weder Instanz noch Klasse – siehe oben), wird beim Aufruf aber automatisch übergeben und ist eine Besonderheit von Python (UML-Diagramme sind nicht auf eine bestimmte Programmiersprache beschränkt). 

## Beispiel: Bibliotheksverwaltung

In diesem Abschnitt wollen wir eine einfache Bibliotheksverwaltung programmieren. Das geplante System wird im folgenden UML-Klassendiagramm dargestellt:   

![UML-Klassendiagramm für die Bibliotheksverwaltung.](images/oop-uml-library.png){#fig-uml-lib width=75%}

### Die Klasse `Library` {.unnumbered}
Das einzige Attribut der Klasse `Library` ist `media_collection` und stellt eine Liste aller Medien im Bestand dar. Über die Methoden der Klasse können Medien entliehen (`borrow_medium(medium_id)`) bzw. zurückgegeben (`return_medium(medium_id)`), dem Bestand hinzugefügt (`add_medium(medium)`) bzw. aus ihm entfernt (`remove_medium(medium_id)`) werden. Schließlich soll `get_media_info()` eine Übersicht über alle Medien der Bibliothek liefern.  

### Die Klasse `Medium` und ihre Kindklassen: `Book` und `Magazine` {.unnumbered}
Die Klasse `Medium` auf der anderen Seite stellt eine Basisklasse für alle in der Bibliothek verfügbaren Medien dar (in unserem Beispiel Bücher und Magazine). Alle Medien besitzen einen Titel und eine eindeutige ID. Außerdem kennzeichnet ein boolesches Attribut, ob sie zurzeit verfügbar sind oder nicht (`available`). Die Methode `get_info()` soll die wesentlichen Informationen über das Medium und den aktuellen Ausleih-Status auf dem Bildschirm ausgeben. Die Klassen `Book` und `Magazine` erben von `Medium` und erweitern die Basisklasse um spezifische Attribute. Die Methode `get_info()` soll jeweils überschrieben werden.

Die Medien-ID hat eine wichtige Bedeutung: Wie wir an den Methoden zum Ausleihen und Zurückgeben erkennen, arbeitet die Bibliothek intern ausschließlich mit dieser eindeutigen Kennzeichnung und nicht etwa mit dem Titel oder anderen Attributen. Sie muss auf jeden Fall eindeutig sein! Es stellt sich noch die Frage, wann und durch wen ein Medium seine ID erhält. In der realen Welt sicherlich durch die Bibliothek, vermutlich in dem Moment, in dem es in den Bestand der Bibliothek aufgenommen wird. Wir wollen das Beispiel aber zunächst einfacher halten und weisen jedem Medium bei seiner Erzeugung händisch eine ID zu. 

### Mit `super()` auf die Basisklasse zugreifen {.unnumbered} 
Bevor wir mit dem Programmieren beginnen, möchten wir noch auf eine wichtige Besonderheit hinweisen:
Da die abgeleiteten Klassen diesmal eigene zusätzliche Attribute besitzen, reicht es nicht aus, einfach den Konstruktor der Basisklasse zu übernehmen. Stattdessen müssen wir den Konstruktor in den Unterklassen selbst definieren.
Dabei können wir jedoch weiterhin den bereits bestehenden Konstruktor der Basisklasse nutzen, um deren Attribute korrekt zu initialisieren. Dafür stellt Python die Funktion `super()` bereit. Mit ihrer Hilfe können wir gezielt Methoden der übergeordneten Klasse aufrufen.
Konkret bedeutet das: Mit `super().__init__(...)` rufen wir den Konstruktor der Basisklasse auf und können ihm die entsprechenden Werte übergeben. 

### Die Implementierung {.unnumbered}
Jetzt sind wir bereit für die Implementierung, wir beginnen auf der rechten Seite mit den Medien:

```python
class Medium:
    
    def __init__(self, title, medium_id):
        self.title = title
        self.medium_id = medium_id
        self.available = True  # wird auf True initialisiert
        
    def get_info(self):
        status = "verfügbar" if self.available else "entliehen"
        print(f"ID: {self.medium_id}, Titel: {self.title}, Status: {status}")

class Book(Medium):

    def __init__(self, title, medium_id, author, isbn):
        # Konstruktor der Basisklasse aufrufen:
        super().__init__(title, medium_id)
        # zusätzliche Attribute zuweisen:
        self.author = author
        self.isbn = isbn
        
    def get_info(self):     
        # Wir könnten auch hier super() verwenden,
        # allerdings könnten wir dann die neuen Attribute 
        # nur darunter anfügen: 
        # super().get_info()
        # print(f"Autor: {self.author}, ISBN: {self.isbn}")
        
        status = "verfügbar" if self.available else "entliehen"
        print(f"ID: {self.medium_id}, Titel: {self.title}, Autor: {self.author}, ISBN: {self.isbn}, Status: {status}")

class Magazine(Medium):

    def __init__(self, title, medium_id, issue, year):
        # Konstruktor der Basisklasse aufrufen:
        super().__init__(title, medium_id)
        # zusätzliche Attribute zuweisen:
        self.issue = issue
        self.year = year
    
    def get_info(self):     
        status = "verfügbar" if self.available else "entliehen"
        print(f"ID: {self.medium_id}, Titel: {self.title}, Ausgabe: {self.issue}, Jahr: {self.year}, Status: {status}") 
```

Legen wir ein paar Beispiel-Medien an:

```python
book1 = Book("Herr der Ringe", 1, "Tolkien, J. R. R.", "978-3-16-148410-0")
magazine1 = Magazine("Nature", 2, 5, 2004)
book1.get_info()
magazine1.get_info()
```

```python
class Library:
    
    def __init__(self):
        self.media_collection = []
        
    def borrow_medium(self, medium_id):
        
        # Bestand wird durchsucht 
        for medium in self.media_collection:
        
            if medium.medium_id == medium_id:
                
                # Das Medium wurde im Bestand gefunden 
                
                if medium.available:
                    medium.available = False
                    print(f"{medium.title} wurde entliehen.")
                else:
                    print(f"AUSLEIHEN NICHT MÖGLICH! {medium.title} ist bereits entliehen.")
                
                return
                
        print("Das gewünschte Medium befindet sich nicht im Bestand.")

    def return_medium(self, medium_id):
        
        for medium in self.media_collection:
            
            if medium.medium_id == medium_id:
                
                if not medium.available:
                    medium.available = True
                    print(f"{medium.title} wurde zurückgegeben.")
                else:
                    print(f"Fehler: {medium.title} wurde noch nicht entliehen.")
                
                return
                
        print("Das Medium gehört nicht zum Bestand der Bibliothek.")

    def add_medium(self, medium):
        
        medium.available = True
        self.media_collection.append(medium)

    def remove_medium(self, medium_id):
        
        for medium in self.media_collection:
            
            if medium.medium_id == medium_id:
                self.media_collection.remove(medium)
                print(f"{medium.title} wurde aus dem Bestand entfernt.")
                return
                
        print("Das Medium gehört nicht zum Bestand der Bibliothek.")

    def get_media_info(self):
        print("Bibliotheks-Bestand:")
        for medium in self.media_collection:
            medium.get_info()
```

```python
# Es folgen ein paar Tests unseres Systems
library = Library()
library.add_medium(book1)
library.add_medium(magazine1)
library.get_media_info()
```

```python
library.borrow_medium(1)
library.get_media_info()
```

```python
library.borrow_medium(1)
library.borrow_medium(2)
library.return_medium(1)
library.get_media_info()
```

## Public, Protected, Private

In objektorientierten Programmiersprachen wie Java oder C++ ist die Sichtbarkeit von Attributen und Methoden streng durch Zugriffsmodifikatoren geregelt: public, protected und private. Python kennt diese Schlüsselwörter nicht in derselben Weise, bietet aber eigene **Konventionen**, um gewünschte Zugriffseinschränkungen auf Klassenbestandteile auszudrücken. 

### public - der Standard {.unnumbered}
In Python ist alles standardmäßig public. Das bedeutet, dass auf Methoden und Attribute von außerhalb der Klasse direkt zugegriffen werden kann. In den bisherigen Beispielen konnte man sowohl auf Attribute (z.B. `issue` in unserer `Magazine`-Klasse) als auch auf Methoden (z.B. `get_info()`) uneingeschränkt zugreifen. Das ist in anderen Programmiersprachen eher die Ausnahme als die Regel.  

### protected - nur für den internen Gebrauch {.unnumbered}

Python verwendet einen führenden Unterstrich (`_`), um zu signalisieren: Dieses Attribut oder diese Methode sollte von außen nicht direkt verwendet werden, sondern nur innerhalb der Klasse und ihrer abgeleiteten Klassen. Technisch ist der Zugriff aber **weiterhin möglich** – es handelt sich nur um eine Konvention, kein Verbot. Hier wird klar, dass Python stark auf die Verantwortung der Programmiererinnen und Programmierer setzt, statt auf strenge Zugriffsverbote. Ein Beispiel:

```python
class Vehicle:
    def __init__(self):
        self._fuel_level = 100  # protected: nur für interne Verwendung gedacht

class Car(Vehicle):
    def show_fuel(self):
        print(f"Aktueller Tankstand: {self._fuel_level}%")
    def refuel(self):
        self._fuel_level = 100
    def drive(self, velocity, distance):
        pass  # Platzhalter für die Verbrauchsberechnung während einer Fahrt  
```

### private - Gehört nur einer einzigen Klasse! {.unnumbered}

Private bedeutet, dass dieses Attribut oder diese Methode nur zu genau einer Klasse gehören soll und nicht zu davon abgeleiteten Klassen. Sie werden durch zwei führende Unterstriche (`__`) gekennzeichnet. Hier handelt es sich um mehr als nur eine Konvention: Python reagiert auf die beiden führenden Unterstriche mit *Name Mangling*: Das heißt, Python verändert intern den Namen und stellt den Namen der *besitzenden* Klasse voran (wiederum mit einem führenden Unterstrich). Kennt man den Mechanismus, könnte man über den neuen Namen wie gewohnt auf das Attribut oder die Methode zugreifen (auch von außerhalb oder innerhalb abgeleiteter Klassen), das ist aber nicht erwünscht. Beispiel: 

```python
class Engine:
    def __init__(self):
        self.__power = 150  # private: gehört nur der Klasse Engine 

    def show_power(self):
        print(f"Leistung: {self.__power} PS")
```

```python
test_engine = Engine()
# So ist es gedacht:
test_engine.show_power()      

# So nicht und es gibt einen Fehler:
# print(test_engine.__power)
# Grund: Intern heißt das Attribut nicht mehr __power, sondern _Engine__power.

print(test_engine._Engine__power)  # funktioniert – aber unsauber!
```

Wichtig: `__power` ist nicht wirklich privat. Es wird intern als `_Engine__power` gespeichert. Über diesen Namen kann man darauf zugreifen, sollte es aber nicht tun. Gedacht ist der doppelte Unterstrich vor allem dafür, Namenskonflikte in abgeleiteten Klassen zu vermeiden. In der Praxis kennzeichnet man nicht-öffentliche Attribute meist mit einem einfachen Unterstrich. In anderen Sprachen heißt `private` auch wirklich `private`: kein Zugriff von außerhalb der Klasse ist möglich (auch nicht aus abgeleiteten Klassen heraus)!

### Getter und Setter {.unnumbered}

Getter- und Setter-Methoden sind einfache Funktionen, die den kontrollierten Zugriff auf protected (`_attribute`) oder private (`__attribute`) Attribute ermöglichen. Die Idee: Über diese Funktionen wird eine öffentliche, kontrollierte Schnittstelle zu den geschützten Attributen angeboten.
    
* **Getter** liefern den aktuellen Wert eines Attributs zurück.

* **Setter** erlauben das Ändern eines Attributs, oft mit Prüfungen oder zusätzlicher Logik, um den gültigen Wertebereich einzuhalten. 

```python
class Vehicle:
    def __init__(self):
        self._fuel_level = 100  # protected Attribut: nur intern direkt verwenden!

    def get_fuel_level(self):  # Getter: zum Abrufen des Attributs von außen.
        return self._fuel_level

    def set_fuel_level(self, value):  # Setter mit Validierung: zum kontrollierten Setzen des Attributs.
        if 0 <= value <= 100:
            self._fuel_level = value
```

Getter und Setter unterstützen das Prinzip der Kapselung und sind in der objektorientierten Programmierung stark verbreitet.

## Magic Methods

Magic Methods – auch bekannt als *dunder methods* (wegen der doppelten Unterstriche: double underscore) – sind spezielle Methoden einer Klasse, die eine ganz bestimmte Bedeutung haben. Sie ermöglichen es selbstprogrammierten Objekten, sich wie eingebaute Python-Objekte zu verhalten, z.B. bei Vergleichen, Rechenoperationen oder der Darstellung im Print-Output. *Magisch* ist daran, dass diese Methoden automatisch von Python aufgerufen werden, wenn bestimmte Operationen auf einem Objekt durchgeführt werden – man muss sie nicht explizit aufrufen. Ein Beispiel ist der Konstruktor, der automatisch bei der Objekt-Erzeugung aufgerufen wird. Es gibt eine Vielzahl von Magic Methods, wir stellen einige vor:  

| Methode    | Wird aufgerufen bei …           | Zweck                                  |
| ---------- | ------------------------------- | -------------------------------------- |
| `__init__` | Objekt-Erzeugung                | Initialisierung des Objekts            |
| `__str__`  | `str(obj)` oder `print(obj)`    | String-Repräsentation für Ausgaben     |
| `__eq__`   | `obj1 == obj2`                  | Vergleich auf Gleichheit               |
| `__lt__`   | `obj1 < obj2`                   | Kleiner-Vergleich                      |
| `__add__`  | `obj1 + obj2`                   | Addition mit benutzerdefinierter Logik |
| `__len__`  | `len(obj)`                      | Länge eines Objekts                    |
| `__int__`  |  `int(obj)`                     | Integer-Repräsentation des Objekts (analog für `float`) | 

Übrigens steckt auch hinter `+=` eine Magic Method: `__iadd__`. Ist sie definiert, verändert `obj += x` das Objekt direkt – so bei Listen (siehe Kapitel *Datenstrukturen*). Fehlt sie, führt Python stattdessen `obj = obj + x` aus und verwendet dafür `__add__`.

Wir erweitern unsere Klasse `Book`, um einige Magic Methods exemplarisch vorzustellen: 

```python
class Book(Medium): 
    def __init__(self, title, medium_id, author, isbn):
        super().__init__(title, medium_id)
        self.author = author
        self.isbn = isbn
    def get_info(self):                
        
        status = "verfügbar" if self.available else "entliehen"
        
        print(f"ID: {self.medium_id}, Titel: {self.title}, Autor: {self.author}, ISBN: {self.isbn}, Status: {status}")

    # der neue Teil: #
    #----------------#

    # wir definieren eine String-Repräsentation für Books:
    def __str__(self):
        return f"'{self.title}' von {self.author}"

    # wir definieren den Gleichheitsoperator für Books:
    def __eq__(self, other):
        if not isinstance(other, Book):
            return False  # ein Book ist nie gleich einem Objekt anderen Typs
        return self.isbn == other.isbn

    # wir definieren eine Integer-Repräsentation für Books:
    def __int__(self):
        return self.medium_id
    
```

### Operatoren überladen {.unnumbered}
Ein **Operator** ist eine Vorschrift, die aus einer Reihe von **Operanden** einen neuen Wert berechnet. Typische Beispiele sind Rechenoperatoren, wie `+` oder `*`, aber auch Vergleichsoperatoren wie `<` oder `==` (bei Letzteren ist das Ergebnis ein Wahrheitswert). Mit Magic Methods kann man in Python **Operatoren überladen** – also ihnen eine benutzerdefinierte Bedeutung für eigene Klassen geben.

Wichtige Magic Methods für Rechenoperationen: 

| Operator   | Magic Method                | Beschreibung                    |
| ---------- | --------------------------- | ------------------------------- |
| `+`        | `__add__(self, other)`      | Addition                        |
| `-`        | `__sub__(self, other)`      | Subtraktion                     |
| `*`        | `__mul__(self, other)`      | Multiplikation                  |
| `/`        | `__truediv__(self, other)`  | Division (Gleitkomma)           |
| `//`       | `__floordiv__(self, other)` | Ganzzahl-Division               |
| `%`        | `__mod__(self, other)`      | Modulo (Rest)                   |
| `**`       | `__pow__(self, other)`      | Potenz                          |   
| `abs(obj)` | `__abs__(self)`             | Absolutwert / Betrag            |

Wichtige Magic Methods für Vergleichsoperationen:

| Operator | Magic Method          | Beschreibung        |
| -------- | --------------------- | ------------------- |
| `==`     | `__eq__(self, other)` | Gleichheit          |
| `!=`     | `__ne__(self, other)` | Ungleichheit        |
| `<`      | `__lt__(self, other)` | Kleiner als         |
| `<=`     | `__le__(self, other)` | Kleiner oder gleich |
| `>`      | `__gt__(self, other)` | Größer als          |
| `>=`     | `__ge__(self, other)` | Größer oder gleich  |

Hinweis: Wenn `__eq__(...)` implementiert wird, nicht aber `__ne__(...)`, funktioniert `!=` automatisch als Negation von `__eq__(...)`.

## Abstrakte Klassen

Abstrakte Klassen sind spezielle Klassen, die man nicht direkt benutzen kann, um Objekte zu erstellen. Stattdessen dienen sie als Vorlage für andere Klassen. Sie definieren Methoden, die alle abgeleiteten Klassen verpflichtend selbst umsetzen müssen. Man könnte sagen: Die abstrakte Klasse gibt die Struktur für alle konkreten Klassen vor. Allerdings handelt es sich dabei um eine Art *Basisstruktur*, jede abgeleitete Klasse kann darüber hinaus weitere Attribute und Methoden anlegen.

Im UML-Klassendiagramm kennzeichnen wir eine abstrakte Klasse durch ein `<<abstract>>` über dem Klassennamen (alternativ kann eine kursive Schrift für den Klassennamen verwendet werden, was aber leichter übersehen werden kann). Abstrakte Methoden werden kursiv gesetzt. Das folgende Beispiel illustriert das Konzept der abstrakten Klassen am Beispiel von geometrischen Figuren: `Shape` ist eine abstrakte Basisklasse mit zwei abstrakten Methoden, nämlich `area()` und `perimeter()` (Flächen- und Umfangsberechnung). Alle konkreten geometrischen Formen werden von dieser Basisklasse abgeleitet und müssen verpflichtend `area()` und `perimeter()` überschreiben. Dadurch wird ein einheitliches Design aller geometrischen Formen sichergestellt. Eine andere Sichtweise ist: Die abstrakte Klasse legt das Interface für die Benutzung der Klassen fest.  

![UML-Klassendiagramm mit abstrakter Basisklasse](images/oop-uml-abstract.png){#fig-uml-abstract width=80%}

In Python nutzt man für die Umsetzung im Code das Modul **abc** (*abstract base classes*, abstrakte Basisklassen). Im folgenden Codebeispiel sehen wir, wie dies funktioniert: 

- Die abstrakte Klasse wird von `ABC` abgeleitet.
- Ihre Methoden werden mit einem sog. Decorator, `@abstractmethod`, als abstrakte Methoden markiert. 

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        """Flächeninhalt berechnen"""
        pass

    @abstractmethod
    def perimeter(self):
        """Umfang berechnen"""
        pass
```

Die abstrakte Klasse `Shape` kann natürlich nichts berechnen (Welche Fläche?! Welchen Umfang?!), sie dient nur als eine Art "Bauplan" für alle konkreten Klassen, die nun von ihr abgeleitet werden:  

```python
import math
class Rectangle(Shape):
    def __init__(self, width, height):
        self._width = width
        self._height = height

    def area(self):
        return self._width * self._height

    def perimeter(self):
        return 2 * (self._width + self._height)

class Circle(Shape):
    def __init__(self, radius):
        self._radius = radius

    def area(self):
        return math.pi * self._radius ** 2

    def perimeter(self):
        return 2 * math.pi * self._radius

class Triangle(Shape):
    def __init__(self, a, b, c):
        self._a = a
        self._b = b
        self._c = c

    def area(self):
        # heronsche Formel
        s = self.perimeter() / 2
        return math.sqrt(s * (s - self._a) * (s - self._b) * (s - self._c))

    def perimeter(self):
        return self._a + self._b + self._c

# Test

shapes = [
    Rectangle(4, 5),
    Circle(3),
    Triangle(3, 4, 5)
]

for shape in shapes:
    print(f"{shape.__class__.__name__}: Fläche = {shape.area():.2f}, Umfang = {shape.perimeter():.2f}")
```

Durch die Verwendung des Moduls `abc` garantieren wir, dass die abstrakten Methoden vollständig implementiert werden müssen. Im Folgenden verletzen wir diese Regel und definieren eine unvollständige Klasse `Hexagon`, die die abstrakte Methode `perimeter` nicht implementiert: 

```python
class Hexagon(Shape):
    def __init__(self, side_length):
        self._side_length = side_length

    def area(self):
        return (3 * math.sqrt(3) / 2) * self._side_length ** 2
```

Beim Versuch, ein `Hexagon` zu instanziieren, erhalten wir folgerichtig eine Fehlermeldung: 

```python
hexagon = Hexagon(3.0)
```
